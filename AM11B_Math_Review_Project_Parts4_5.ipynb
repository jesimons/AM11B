```python
# 1. Define the multivariable function g(s,t)
def g(s, t):
    return s**3 + t   # EDIT HERE

# 2. Create a grid of values for s and t
s_vals = np.linspace(-2, 2, 100) # MAY EDIT HERE (s goes from -2 to 2)
t_vals = np.linspace(-5, 5, 100) # MAY EDIT HERE (t goes from -5 to 5)
S, T = np.meshgrid(s_vals, t_vals)

# 3. Calculate the corresponding z values (g(s,t))
Z = g(S, T)

# 4. Create the plot with two subplots
fig = plt.figure(figsize=(14, 6))

# Subplot 1: 2D Level Curves (Contour Plot)
ax1 = fig.add_subplot(121) # 1 row, 2 columns, 1st subplot

# Define the levels for the contour lines
levels = np.linspace(Z.min(), Z.max(), 15) # Adjust number of levels as needed

contour = ax1.contour(S, T, Z, levels=levels, cmap='viridis')
ax1.set_xlabel('s')
ax1.set_ylabel('t')
ax1.set_title('Level Curves of g(s,t)')
fig.colorbar(contour, ax=ax1, label='g(s,t) value')
#ax1.set_aspect('equal', adjustable='box') # Make the aspect ratio equal for better visualization

# Subplot 2: 3D Surface Plot
ax2 = fig.add_subplot(122, projection='3d') # 1 row, 2 columns, 2nd subplot

surface = ax2.plot_surface(S, T, Z, cmap='viridis', edgecolor='none')
ax2.set_xlabel('s')
ax2.set_ylabel('t')
ax2.set_zlabel('g(s,t)')
ax2.set_title('3D Surface Plot of g(s,t)')
fig.colorbar(surface, ax=ax2, shrink=0.5, aspect=5, label='g(s,t) value')

plt.tight_layout()
plt.show()
```

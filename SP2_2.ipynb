#

*2.* Apply the Fundamental Theorem of Calculus Part 1
#
#
The FTC Part 1 states:
If we have $\frac{d}{dx}\int_a^x f(t) dt$, then the derivative is simply $f(x)$.


In this problem, $f(t) = 1$, so when we replace $t$ with $x$, we get $f(x) = 1$.

Therefore:

$$\frac{d}{dx}\int_0^x 1 dt = 1$$

**3.** Verify (optional)

$$\int_0^x 1 dt = [t]_0^x = x - 0 = x$$

Now take the derivative:

$$\frac{d}{dx}(x) = 1$$

**Answer:** $1$

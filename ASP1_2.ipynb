**Problem 2:**

Suppose a country has the Lorenz curve

$L(x) = x^2$,

where $x$ is the cumulative share of the population (from 0 to 1).

a) Calculate the Gini coefficient.  
b) What does this value tell us about income inequality?  
c) What percentage of total income is earned by the bottom 50% of the population?

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**a) Calculating the Gini coefficient**

The formula for the Gini coefficient is

$G = 1 − 2 \int_0^1 L(x) dx$

Here $L(x) = x^2$, so

$\int_0^1 L(x) dx = \int_0^1 x^2 dx$
= [$ x^3 / 3 $ ] from 0 to 1
= 1/3

So

$G = 1 − 2 · (1/3)$
= $1 − 2/3$
= $1/3 ≈ 0.33$

**b) Interpretation**

A Gini coefficient of about 0.33 means the country has **moderate / relatively low inequality**.  
Income is not perfectly equal, but it is not extremely unequal either.

**c) Income of the bottom 50%**

The Lorenz curve $L(x)$ tells us the share of income earned by the bottom x of the population.

For the bottom 50% ($x = 0.5$):

$L(0.5) = (0.5)^2 = 0.25$

So the bottom 50% of the population earns **25% of total income**.

**Problem 3:**

*The small coastal town of Harbor Cove wants to understand income inequality among its residents. A community research group surveys household incomes and constructs the following Lorenz curve, which represents the cumulative share of total income earned by the poorest households: $L(x) = 0.4x^2 + 0.6x^3$, where x represents the cumulative share of households (from 0 to 1) and L(x) represents the cumulative share of income earned. The town council wants to know how unequal income is in Harbor Cove, and need the Gini coefficient to do so.*

*a) Calculate the Gini coefficient.*

*b) What does this value tell us about income inequality?*

*c) What percentage of income is earned by the bottom 50% [this value can change] of the population?*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Calculating G**

Using the formula:
$$G = 1 - 2\int_0^1 L(x) \, dx$$

*Input values into the integral:*

*$L(x) = 0.4x^2 + 0.6x^3$*

*$G = 1 - 2\int_0^1 [0.4x^2 + 0.6x^3] \, dx$*

*Separate the integral and evaluate them from 0 to 1:*

*$G = 1 - 2\int_0^1 [0.4x^2 + 0.6x^3] \, dx = \int_0^1 0.4x^2 \, dx + \int_0^1 0.6x^3 \, dx$*

*$\int_0^1 0.4x^2 \, dx = \frac{[0.4x^3]_0^1}{3} = \frac{0.4(1)^3}{3} - \frac{0.4(0)^3}{3} = \frac{0.4}{3}$*

*$\int_0^1 0.6x^3 \, dx = \frac{[0.6x^4]_0^1}{4} = \frac{0.6(1)^4}{4} - \frac{0.6(0)^4}{4} = \frac{0.6}{4}$*

*Add them together:*

*$\int L(x) \, dx = \frac{0.4}{3} + \frac{0.6}{4} = \frac{1.6}{12} + \frac{1.8}{12} = \frac{3.4}{12}$*

*Plug into the Gini formula:*

*$G = 1 - 2\frac{3.4}{12} = 1 - \frac{6.8}{12} = \frac{5.2}{12} ≈ 0.433$*
    
**Part (b): Interpretation**

*Gini coefficient is interpreted on a scale of 0 to 1, with 0 being "perfect equality" of wealth and 1 being "perfect inequality" of wealth. A Gini coefficient of 0.43 means that Harbor Cove has moderate income inequality, where there is the existence of a middle class but the upper class holds a larger portion of the wealth than lower income households.*

**Part (c): Income Share**

*Evaluate F(0.5) by plugging in x = 0.5 to the L(x) function:*

*$L(x) = 0.4x^2 + 0.6x^3$*

*$L(0.5) = 0.4(0.5)^2 + 0.6(0.5)^3 = 0.4(0.25) + 0.6(0.125) = 0.1 = 0.075 = 0.175$*

*$L(0.5) = 0.175 = 17.5\%$*

*The bottom 50% of Harbor Cove households earn 17.5% of the total income, while the top 50% earns 82.5%. This indicates a notable degree of income inequality.*

**Problem 6:**

**Suppose the Lorenz curve representing the income inequality in The Gambia is modeled by:**

$$L(x)=xe^{4(x-1)}$$
  
*a) Calculate the Gini coefficient.*

*b) What does this value tell us about income inequality?*

*c) What percentage of income is earned by the bottom 50% [this value can change] of the population? What about the top 1% of the population?*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Calculating G**

*The Gini coefficient implied by a Lorenz curve is equal to one minus twice the area under the Lorenz curve. Set up the equation, including the integral, integrate, evaluate, and simplify.*

$$G=1-2\int_0^1 xe^{4(x-1)}\,dx$$

$$u=p,dv=4(p-1)$$

$$du=dp$$

$$w=4(p-1)$$

$$dw=4$$

$$v=\int e^{4(p-1)}\,dp=\int e^w\,1/4dw$$

$$v=1/4\int e^w\,dw$$

$$v=1/4e^w$$

$$v=1/4e^{4(p-1)}\,dw$$

$$G=1-2(uv-\int_0^1 vdu)$$

$$G=1-2(1/4pe^{4(p-1)}-\int_0^1 1/4e^{4(p-1)}\,dp)$$

$$G=1-2(1/4pe^{4(p-1)}-1/4\int_0^1 e^{4(p-1)}\,dp)$$

$$G=1-2(1/4pe^{4(p-1)}-1/16e^{4(p-1)}|_0^1)$$

$$G=1-2(1/4(1)e^{4(1-1)}-1/16e^{4(1-1)}-(1/4(0)e^{4(0-1)}-1/16e^{4(0-1)}))$$

$$G=1-2(1/4e^0-1/16e^0-(1/4(0)e^{-4}-1/16e^{-4}))$$

$$G=1-2(1/4-1/16+1/16e^{-4})$$

$$G=1-2(3+e^{-4}/16)$$

$$G=1-(6+2e^{-4}/16))$$

$$G=(16-6-2e^{-4})/16$$

$$G=(10-2e^{-4})/16$$

$$G=(5-e^{-4})/8$$

$$G≈0.6227$$

**Part (b): Interpretation**

**0.6227 is a relatively high Gini coefficient relative to most countries, indicating a very unequal income distribution. The U.S., which is notably less income-equal than most other industrialized wealthy countries, has a Gini coefficient in the low 0.40s. Sweden, a wealthy country with a robust welfare state and high taxes, has a Gini of about 0.30. Brazil has a relatively high Gini of about 0.50. South Africa, perhaps the most income-unequal country in the world, has a Gini in the low 0.60s just like The Gambia in this hypothetical example.

**Part (c): Income Share**

*Evaluate L(x) at x=c to find the share of income received by the bottom c%.

$$L(0.5)=0.5e^{4(0.5-1)}$$

$$L(0.5)=0.5e^-2$$

$$L(0.5)≈0.0677$$

$$0.0677*100=6.77$$

**This means the bottom half of the population earns only about 6.77% of the income.**

$$1-L(0.9)=1-0.9e^{4(0.9-1)}$$

$$1-L(0.9)=1-0.9e^{-0.4}$$

$$1-L(0.9)≈0.3967$$

$$0.3967*100=39.67$$

**This means the top 10% of the population earns almost 40 of the income.**

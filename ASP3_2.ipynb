## Sample Problems 3.2: Price-Elasticity of Demand

These problems are applications of derivatives in econ. A few notes:

**Price-Elasticity of Demand** refers to a ratio of percentage changes in demand to prices. For instance, if you raise the price of a burrito by 10%, and the demand changes by 20%, the price-elasticity of demand is:
$$\frac{20 \%}{10 \%} = 2$$
The higher this ratio, the more sensitive the market is to price changes. The lower this ratio, the less sensitive (e.g. less **elastic**) the market is.  

When a product is a necessity (like food), elasticity tends to be low. Grocery staples like milk have low price elasticity. For example,  if the price of a gallon of milk rises from $ \$3.50$ to $ \$3.70$, most families still buy about the same amount because they consider it essential.


If a specialty snack gets more expensive at a regular supermarket, a shopper might buy a smaller amount, wait to find it cheaper at Grocery Outlet, or skip it entirely. For example, a shopper usually buys a specialty snack at Grocery Outlet for $ \$2.00$ a bag and picks up four bags a month. When they see the same snack priced at $ \$5.50$ at a regular supermarket, they buy none and wait to get it at Grocery Outlet.


Since we have used the word "change" above, we can capture this concept with this formula for elasticity:
$$ E_{q/p} = \frac{\% \Delta q}{\% \Delta p}= \frac{\frac{\Delta q}{q}}{\frac{\Delta p}{p}} = \frac{\Delta q}{\Delta p}\frac{p}{q}$$
The formula above is written for **average elasticity**. To bring this into the realm of calculus, we have the formula:

$$ \eta_{q/p} =  \frac{d q}{d p}\frac{p}{q}$$

This formula above is called **point elasticity** and is the elasticity formula we will use.

**Problem 2:**

A Pop Mart store models the quantity demanded ($q$, in thousands of units) of a new limited gold edition Labubu, based on its price ($p$ in dollars), using the following demand function:$$q(p) = \frac{1000}{p^2}$$Find the price elasticity of demand ($\eta_{q/p}$) when the price is $p = \$5$.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

1. State the Elasticity Formula

The formula for the price elasticity of demand is:
$$\eta_{q/p} = \frac{p}{q} \cdot \frac{dq}{dp}$$

2. Calculate $q$ at $p=5$

Substitute $p=5$ into the demand function $q(p) = \frac{1000}{p^2}$:
$$q(5) = \frac{1000}{(5)^2} = \frac{1000}{25} = \mathbf{40}$$

3. Calculate the Derivative $\frac{dq}{dp}$

First, rewrite the demand function: $q(p) = 1000p^{-2}$.
Differentiate with respect to $p$:
$$\frac{dq}{dp} = \frac{d}{dp}(1000p^{-2}) = -2000p^{-3} = -\frac{2000}{p^3}$$
Evaluate the derivative at $p=5$:
$$\frac{dq}{dp}\bigg|_{p=5} = -\frac{2000}{(5)^3} = -\frac{2000}{125} = \mathbf{-16}$$

4. Calculate the Elasticity $\eta_{q/p}$ at $p=5$

Substitute the calculated values into the elasticity formula:
$$\eta_{q/p}\bigg|_{p=5} = \frac{p}{q} \cdot \frac{dq}{dp} = \frac{5}{40} \cdot (-16)$$
$$\eta_{q/p}\bigg|_{p=5} = \frac{1}{8} \cdot (-16) = \mathbf{-2}$$

5. Interpretation

The price elasticity of demand is $\mathbf{-2}$. In absolute terms, $|\eta_{q/p}| = 2$. This means that if Pop Mart were to raise the price of the Labubu figure by $1\%$, the quantity demanded would decrease by approximately $\mathbf{2\%}$. Since $|\eta_{q/p}| > 1$, demand for the figure is elastic at the $\$5$ price point.

*The elasticity is constant: $E(p) = -1.2$ for every $p > 0$.*

*Since $p = 10$ and $10 > 0$, $E(10) = -1.2$.*

**Problem 12:**

In the market for coffee, firms face the demand curve:

$$q(p)=4p^2e^{-p}$$

Find the price elasticity of demand at p=3.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

$$\eta_{q/p}=\frac{d q}{d p}\frac{p}{q}$$

$$\frac{d q}{d p}=4(2p*e^{-p}+p^2*e^{-p}*-1)$$
$$\frac{d q}{d p}=4(2pe^{-p}-p^2e^{-p})$$
$$\frac{d q}{d p}=4e^{-p}(2p-p^2)$$

$$ \eta_{q/p}=4e^{-p}(2p-p^2)*\frac{p}{4p^2e^{-p}}$$
$$ \eta_{q/p}=\frac{4e^{-p}(2p-p^2)p}{4p^2e^{-p}}$$
$$ \eta_{q/p}=\frac{(2-p)p}{p}$$
$$ \eta_{q/p}=2-p$$

$$ \eta_{q/3}=2-3$$
$$ \eta_{q/3}=-1$$
$$ |\eta_{q/3}|=1$$

This means that when the price is at $3, a change in price will cause a change in the quantity of coffee demanded that is similar in proportion.

</details>

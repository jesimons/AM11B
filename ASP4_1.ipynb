**Problem 2:**

A small tech firm produces specialized circuit boards using labor ($L$, in worker-hours) and capital ($K$, in machine-hours). Their total output, $Q$, is modeled by the production function:

$$Q(L, K) = 50L^{0.6}K^{0.4}$$

Derive the Marginal Product of Labor ($MP_L$) and the Marginal Product of Capital ($MP_K$).
Evaluate the marginal products when the firm is operating with $L = 100$ worker-hours and $K = 50$ machine-hours.
Interpret the economic meaning of the $MP_L$ result.


**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>


1. Derive the Marginal Product Functions

Marginal Product of Labor ($MP_L$)
We treat $K$ as a constant and differentiate $Q$ with respect to $L$:
$$MP_L = \frac{\partial Q}{\partial L} = \frac{\partial}{\partial L}(50L^{0.6}K^{0.4})$$
$$MP_L = 50K^{0.4} \cdot (0.6 L^{0.6-1})$$
$$\mathbf{MP_L = 30 L^{-0.4} K^{0.4}}$$

Marginal Product of Capital ($MP_K$)
We treat $L$ as a constant and differentiate $Q$ with respect to $K$:
$$MP_K = \frac{\partial Q}{\partial K} = \frac{\partial}{\partial K}(50L^{0.6}K^{0.4})$$
$$MP_K = 50L^{0.6} \cdot (0.4 K^{0.4-1})$$
$$\mathbf{MP_K = 20 L^{0.6} K^{-0.6}}$$

2. Evaluate at $L=100$ and $K=50$

Marginal Product of Labor
$$MP_L(100, 50) = 30 (100)^{-0.4} (50)^{0.4}$$
$$MP_L(100, 50) = 30 \left(\frac{50}{100}\right)^{0.4} \cdot 100^{-0.4} \cdot 100^{0.4}$$
$$MP_L(100, 50) = 30 (0.5)^{0.4} \approx \mathbf{22.73}$$

Marginal Product of Capital
$$MP_K(100, 50) = 20 (100)^{0.6} (50)^{-0.6}$$
$$MP_K(100, 50) = 20 \left(\frac{100}{50}\right)^{0.6}$$

Economic Interpretation

The result $MP_L(100, 50) \approx 22.73$ means that if the firm hires one more worker-hour (increasing $L$ from $100$ to $101$) while holding capital ($K=50$) constant, the total output ($\mathbf{Q}$) will increase by approximately $\mathbf{22.73}$ circuit boards.

*a) Find the marginal product of labor $MP_L = \frac{\partial Q}{\partial L}$ and the marginal product of capital $MP_K = \frac{\partial Q}{\partial K}$.*

*b) Evaluate $MP_L$ and $MP_K$ at $L = 25$ workers and $K = 100$ machines.*

*c) Interpret the results economically.*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Find $MP_L$ and $MP_K$**

*$MP_L = \frac{\partial Q}{\partial L} = 10⋅0.6L^{0.6−1}K{0.4} = 6L^{−0.4}K^{0.4}$*

*$MP_K = \frac{\partial Q}{\partial K} = 10⋅0.4L^{0.6}K^{0.4−1} = 4L^{0.6}K^{−0.6}$*

**Part (b): Evaluate at $L = 25$ and $K = 100$**

*$L^{−0.4} = 25^{−0.4} ≈ 0.27595$*

*$K^{0.4} = 100^{0.4} ≈ 6.30957$*

*$L^{0.6} = 25^{0.6} ≈ 6.89865$*

*$K^{−0.6} = 100^{−0.6} ≈ 0.06310$*

*$MP_L(25,100) = 6(25^{−0.4})(100^{0.4}) ≈ 10.44661$*

*$MP_K(25,100) = 4(25^{0.6})(100^{−0.6}) ≈ 1.74110$*

**Part (c): Economic Interpretation**

*$MP_L(25,100) ≈ 10.45$, meaning that hiring one extra worker at the current output levels ($K = 100$ fixed) increases expected output by 10.45 paint tubes per period.*

*$MP_K(25,100) ≈ 1.74$, meaning that adding one extra machine at the current output levels ($L = 25$ fixed) increases expected output by 1.74 paint tubes per period.*

**Problem 3:**

A small bakery produces loaves of bread using ovens (capital) $K$ and workers (labor) $L$.  
Its production function is

$$Q(K,L) = 10 K^{0.5} L^{0.5},$$

where $Q$ is the number of loaves produced per day.

1. Find the marginal product of labor $\dfrac{\partial Q}{\partial L}$ and the marginal product of capital $\dfrac{\partial Q}{\partial K}$.
2. Evaluate both marginal products at $K = 16$ ovens and $L = 25$ workers.
3. Interpret the results in words (e.g., “At the current production level, hiring one more worker would increase output by approximately __ units”).

*1. Find $\partial Q / \partial L$ and $\partial Q / \partial K$.*

Given  
$$Q(K,L) = 10 K^{0.5} L^{0.5}.$$

Treat $K$ as constant and differentiate with respect to $L$:
$$
\frac{\partial Q}{\partial L}
= 10 K^{0.5} \cdot 0.5 L^{-0.5}
= 5 K^{0.5} L^{-0.5}.
$$

Treat $L$ as constant and differentiate with respect to $K$:
$$
\frac{\partial Q}{\partial K}
= 10 \cdot 0.5 K^{-0.5} L^{0.5}
= 5 K^{-0.5} L^{0.5}.
$$

2. Evaluate at $K = 16$, $L = 25$.

Recall $\sqrt{16} = 4$ and $\sqrt{25} = 5$.

$$
\frac{\partial Q}{\partial L}(16,25)
= 5 \frac{\sqrt{16}}{\sqrt{25}}
= 5 \cdot \frac{4}{5}
= 4.
$$

$$
\frac{\partial Q}{\partial K}(16,25)
= 5 \frac{\sqrt{25}}{\sqrt{16}}
= 5 \cdot \frac{5}{4}
= \frac{25}{4} = 6.25.
$$

3. Interpretation.

At the current production level $(K = 16, L = 25)$:

- $\dfrac{\partial Q}{\partial L} \approx 4$ means hiring one more worker (holding ovens fixed at 16) would increase output by about **4 loaves of bread per day**.  
- $\dfrac{\partial Q}{\partial K} \approx 6.25$ means adding one more oven (holding workers fixed at 25) would increase output by about **6.25 loaves per day**.

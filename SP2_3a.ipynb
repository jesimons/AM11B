**Problem 1:**

$\int x^2(x^3 + 1)^4 dx$

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1:** Choose u = $x^3 + 1$

**Step 2:** Find du = $3x^2 du$      = $x^2$ =  $\frac {1}{3} du$

**Step 3:** Rewrite the integral: $\int x^2 (x^3 + 1)^4 dx$ = $\int (u^4)(\frac {1}{3} du)$ = $\frac {1}{3} \int u^4 du$

**Step 4:** Integrate: $\frac {1}{3} \int u^4 du$ = $(\frac {1}{3}) \frac {u^5}{5} + C$ = $\frac {u^5}{15} + C$

**Step 5:** Substitute back: $\frac {(x^3 + 1)^5}{15} + C$

</details>

**Problem:**
  
$$ \int 2x \sqrt{x^2 + 1} \ dx\ $$

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1:** Choose a suitable substitution for  $u$  
$$ u = x^2 + 1 $$

**Step 2:** Find $du$  

$$ du = 2x \ dx \quad \Rightarrow \quad dx = \frac{du}{2x} $$

**Step 3:** Rewrite the integral in terms of  $u$ and $du$  
    
$$ \int 2x \sqrt{x^2 + 1} \ dx  = \int 2x \sqrt{u} \cdot \frac{du}{2x}  = \int \sqrt{u} \, du $$

**Step 4:** Integrate with respect to  $u$  
    
$$ \int u^{1/2} \ du = \frac{u^{3/2}}{3/2} = \frac{2}{3} u^{3/2} $$

**Step 5:** Substitute back into the original expression in terms of $x$
                   
$$ \frac{2}{3} (x^2 + 1)^{3/2} $$

**Step 6:** Final Answer  
                   
$$ \boxed{\frac{2}{3} (x^2 + 1)^{3/2} }$$

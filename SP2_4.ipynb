**Problem 1:**

Find the area between $y = x^2$ and $y = 2x + 3$
  
**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1:** Find intersection points
$x^2 = 2x +3$

$0 = x^2 - 2x - 3$
$(x-3)(x+1) = 0$

$x=3$ and $x=-1$

**Step 2:** Determine which function is on top

Choose a number between -1 and 3. For this case, we will use 0 and plug that into both equations.

$0^2= 0$                and            $2(0) + 3 = 3$

Since $3 > 0$, the upper function will be $y = 2x + 3$

    
**Step 3:** Set up the integral

**Area**= A= $\int_{-1}^3 (Top - Bottom)dx$

= $\int_{-1}^3 ((2x + 3) - x^2)dx$

= $\int_{-1}^3 (2x + 3)dx$ - $\int_{-1}^3 x^2dx$

= $\int (2x + 3)dx$ = $x^2 + 3x$

= $\int x^2 dx$ = $\frac {x^3}{3}$

= $F(x)$ = $x^2 + 3x - \frac{x^3}{3}$


**Step 4:** Evaluate from -1 to 3

$F(3)$ = $3^2 + 3(3) - \frac {3^3}{3}$ = $9 + 9 - 9 = 9$

$F(-1)$ = $-1^2 + 3(-1) - \frac {-1^3}{3}$ = $1 - 3 + \frac {1}{3}$ = $\frac {-5}{3}

$A$= $F(3) - F(-1)$ = $9  + \frac {5}{3}$

**A**= $\frac {32}{3}$



</details>

**Problem 11:**

*Find the area of the region enclosed by*  
$$ y = x^2 \quad\text{and}\quad y = x + 2. $$

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1: Find intersection points**

Solve $$ (x^2 = x + 2) $$:

$$ x^2 - x - 2 = 0 \quad\Rightarrow\quad (x-2)(x+1)=0 $$

So the curves intersect at $$ (x=-1) $$ and $$ (x=2) $$.

**Step 2: Determine which function is on top**

Test a point between (-1) and \(2\), e.g. \(x=0\):

$$ y_{\text{parabola}} = 0^2 = 0,\qquad y_{\text{line}} = 0 + 2 = 2. $$

Since $$ (2 > 0) $$, the line $$ (y=x+2) $$ is the top function and the parabola $$ (y=x^2) $$ is the bottom on $$ ([-1,2]) $$.

**Step 3: Set up the integral**

Area:
$$ A = \int_{-1}^{2} \bigl( (x+2) - x^2 \bigr)\,dx. $$

**Step 4: Evaluate**

Compute an antiderivative:
$$ \int \bigl(x+2 - x^2\bigr)\,dx = \frac{1}{2}x^2 + 2x - \frac{1}{3}x^3 + C. $$

Evaluate from (-1) to (2):

At $$ (x=2) $$:
$$ \frac{1}{2}(2)^2 + 2(2) - \frac{1}{3}(2)^3
= 2 + 4 - \frac{8}{3}
= 6 - \frac{8}{3} = \frac{10}{3}. $$

At \(x=-1\):
$$ \frac{1}{2}(-1)^2 + 2(-1) - \frac{1}{3}(-1)^3
= \frac{1}{2} - 2 + \frac{1}{3}
= \frac{3}{6} - \frac{12}{6} + \frac{2}{6}
= -\frac{7}{6}. $$

Subtract:
$$ A = \frac{10}{3} - \left(-\frac{7}{6}\right)
= \frac{10}{3} + \frac{7}{6}
= \frac{20}{6} + \frac{7}{6}
= \frac{27}{6}
= \frac{9}{2}. $$


**Final Answer:**

$$ \boxed{A = \frac{9}{2}} $$

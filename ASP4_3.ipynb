$$ \eta_{Y/K} =  \frac{\partial Y}{\partial K}\frac{K}{Y}$$
where $Y$ is output (often called $Q$). These problems are asking you to compute this quantity for given models.

**Problem 2:**
A Foxconn factory assembles iPhones, with total output ($Q$) modeled by the Cobb-Douglas production function:
$$Q(L, K) = 10 L^{0.7} K^{0.3}$$
where $L$ is labor (in thousands of assembly hours) and $K$ is capital (in thousands of automated machine-hours).

Derive the Marginal Product of Labor ($\text{MP}_L$) function.
Derive the formula for the Output Elasticity of Labor ($\eta_{Q/L}$) and simplify it completely.
Interpret the economic meaning of the simplified $\eta_{Q/L}$ result.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

1. Derive the Marginal Product of Labor ($\text{MP}_L$)

We take the partial derivative of $Q$ with respect to $L$, treating $K$ as a constant:
$$\text{MP}_L = \frac{\partial Q}{\partial L} = \frac{\partial}{\partial L}(10 L^{0.7} K^{0.3})$$
$$\text{MP}_L = 10 K^{0.3} \cdot (0.7 L^{0.7-1})$$
$$\mathbf{\text{MP}_L = 7 L^{-0.3} K^{0.3}}$$

2. Derive the Output Elasticity of Labor ($\eta_{Q/L}$)

The point elasticity formula is:
$$\eta_{Q/L} = \frac{\partial Q}{\partial L} \cdot \frac{L}{Q}$$

Substitute $\text{MP}_L$ and the original production function $Q$:
$$\eta_{Q/L} = \left(7 L^{-0.3} K^{0.3}\right) \cdot \frac{L}{10 L^{0.7} K^{0.3}}$$

Simplify the expression by canceling terms ($\frac{K^{0.3}}{K^{0.3}} = 1$) and combining $L$ terms ($L^{-0.3} \cdot L^1 = L^{0.7}$):
$$\eta_{Q/L} = \frac{7}{10} \cdot \left(\frac{L^{0.7}}{L^{0.7}}\right) \cdot 1$$
$$\mathbf{\eta_{Q/L} = 0.7}$$

3. Economic Interpretation

The result $\mathbf{\eta_{Q/L} = 0.7}$ means that a $\mathbf{1\%}$ increase in labor (assembly hours) will result in exactly a $\mathbf{0.7\%}$ increase in the total production of iPhones, holding capital (automated machines) constant. This is the definition of the output elasticity of labor, $\mathbf{\beta}$, for this Cobb-Douglas function.


**Problem 2a:**

*A factory's output $Q$ is described by the constant-elasticity-of-substitution production function $Q = A(aK^ρ+bL^ρ)^{1/ρ}$, where $K$ is capital, $L$ is labor, $A>0$, $a,b>0$ and ρ≠0.*

*a) Find $η_{Q/K}$ and $η_{Q/L}$.*

*b) Evaluate numerically for the paramater values $A=1, a=0.6, b=0.4, ρ=0.5, K=100, and L=25$.*

*c) Interpret the economic meaning.*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Find $η_{Q/K}$ and $η_{Q/L}$.**

*First, differentiate Q with respect to K and with respect to L seperately.*

*$\frac{\partial Q}{\partial K} = A⋅\frac{1}{ρ}(aK^ρ+bL^ρ)^{\frac{1}{p}-1}⋅\frac{d}{dK}(aK^ρ+bL^ρ) = A⋅\frac{1}{ρ}(aK^ρ+bL^ρ)^{\frac{1}{p}-1}⋅aρK^{ρ−1} = AaK^{ρ−1}(aK^ρ+bL^ρ)^{\frac{1}{p}-1}$*

*$\frac{\partial Q}{\partial L} = A⋅\frac{1}{ρ}(aK^ρ+bL^ρ)^{\frac{1}{p}-1}⋅\frac{d}{dL}(aK^ρ+bL^ρ) = AbL^{p-1}(aK^ρ+bL^ρ)^{\frac{1}{p}-1}$*

*Next, sub values into point-elasticity formula.*

*$η_{Q/K} = \frac{\partial Q}{\partial K}⋅\frac{K}{Q} = AaK^{ρ−1}(aK^ρ+bL^ρ)^{\frac{1}{p}-1}⋅\frac{K}{A(aK^ρ+bL^ρ)^{1/ρ}} = \frac{aK^ρ}{aK^ρ + bL^ρ}$*

*$η_{Q/K} = \frac{\partial Q}{\partial L}⋅\frac{L}{Q} = b⋅\frac{L^ρ}{aK^ρ + bL^ρ} = \frac{bL^ρ}{aK^ρ + bL^ρ}$*

**Part (b): Numerical evaluations**

*$A=1, a=0.6, b=0.4, ρ=0.5, K=100, and L=25$*

*First, compute $K^ρ$ and $L^ρ$.*

*$K^ρ = 100^{0.5} = 10$*

*$L^ρ = 25^{0.5} = 5$*

*Now compute the denominator $aK^ρ+bL^ρ$.*

*$aK^ρ+bL^ρ = (0.6)(10) + (0.4)(5) = 6 + 2 = 8$*

*Lastly, compute the elasticity.*

*$η_{Q/K} = \frac{(0.6)(10)}{8} = 0.75$*

*$η_{Q/L} = \frac{(0.4)(5)}{8} = 0.25$*

**Part (b): Economic Interpretation**

*$η_{Q/K} = 0.75$ means that a 1% raise in capital $K$ raises output by 0.75% when $L$ remains constant.*

*$η_{Q/L} = 0.25$ means that a 1% raise in labor $L$ raises output by 0.25% when $K$ remains constant.*

</details>

A firm’s output depends on capital $K$ and labor $L$ according to the production function  

$$Y(K,L) = 2\sqrt{K} + \sqrt{L} = 2K^{1/2} + L^{1/2}.$$

1. Derive the output elasticity of capital:
   $$\eta_{Y/K} = \frac{\partial Y}{\partial K}\cdot\frac{K}{Y}.$$
2. Derive the output elasticity of labor:
   $$\eta_{Y/L} = \frac{\partial Y}{\partial L}\cdot\frac{L}{Y}.$$
3. Evaluate both elasticities at $K = 25$ and $L = 9$.
4. Interpret these numbers economically.



**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

*1. Output elasticity of capital.*

Given  
$$Y(K,L) = 2K^{1/2} + L^{1/2},$$  
take the partial derivative w.r.t. $K$:

$$
\frac{\partial Y}{\partial K}
= 2 \cdot \frac{1}{2} K^{-1/2}
= K^{-1/2}
= \frac{1}{\sqrt{K}}.
$$

So the elasticity of output with respect to capital is

$$
\eta_{Y/K}
= \frac{\partial Y}{\partial K}\cdot\frac{K}{Y}
= \frac{1}{\sqrt{K}}\cdot\frac{K}{Y}
= \frac{\sqrt{K}}{Y}.
$$

*2. Output elasticity of labor.*

Take the partial derivative w.r.t. $L$:

$$
\frac{\partial Y}{\partial L}
= \frac{1}{2} L^{-1/2}
= \frac{1}{2\sqrt{L}}.
$$

So the elasticity of output with respect to labor is

$$
\eta_{Y/L}
= \frac{\partial Y}{\partial L}\cdot\frac{L}{Y}
= \frac{1}{2\sqrt{L}}\cdot\frac{L}{Y}
= \frac{\sqrt{L}}{2Y}.
$$

*3. Evaluate at $K = 25$, $L = 9$.*

First compute output:

$$
Y(25,9) = 2\sqrt{25} + \sqrt{9} = 2\cdot 5 + 3 = 13.
$$

Then

$$
\eta_{Y/K}(25,9)
= \frac{\sqrt{25}}{13}
= \frac{5}{13}
\approx 0.38,
$$

$$
\eta_{Y/L}(25,9)
= \frac{\sqrt{9}}{2\cdot 13}
= \frac{3}{26}
\approx 0.12.
$$

*4. Interpretation.*

At $K = 25$ and $L = 9$:

- $\eta_{Y/K} \approx 0.38$ means a 1% increase in capital raises output by about **0.38%**, holding labor fixed.  
- $\eta_{Y/L} \approx 0.12$ means a 1% increase in labor raises output by about **0.12%**, holding capital fixed.

So at this production level, output is more responsive to changes in capital than to changes in labor.

</details>

**Problem 4:**

A firm has a production function:
Y=4K^1/2 + 2L^2/3

Compute MPK and MPL

Find expressions for the output elasticity of capital and labor

Evaluate at K=25 and L = 8

Interpret economically

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

Production function: Y = 4K^(1/2) + 2L^(2/3)

Marginal Products:
MPK = 2K^(−1/2)
MPL = (4/3)L^(−1/3)

At K = 25, L = 8:
Y = 4(25^(1/2)) + 2(8^(2/3)) = 4(5) + 2(4) = 20 + 8 = 28

Elasticity of capital:
η_Y(K) = MPK · (K/Y)
= [2 · 25^(−1/2)] · (25/28)
= (2/5) · (25/28)
= 10/28 ≈ 0.36

Elasticity of labor:
η_Y(L) = MPL · (L/Y)
= [(4/3) · 8^(−1/3)] · (8/28)
= (4/3)(1/2) · (8/28)
= (2/3) · (8/28)
= 16/84 ≈ 0.19

Interpretation:
A 1% increase in capital increases output by about 0.36%.
A 1% increase in labor increases output by about 0.19%.



</details>

**Problem 8:**

Output in the economy modeled by the function:

$$Y=A(\alpha K^ρ+(1-\alpha)L^ρ)^{1/ρ}$$

Find the output elasticities of labor and capital.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

$$Z=\alpha K^ρ+(1-\alpha)L^ρ$$
$$Y=AZ^{1/ρ}$$

$$\frac{\partial Y}{\partial K}=A(1/ρ)Z^{(1/ρ)-1}* \alpha ρ K^{ρ-1}$$
$$\frac{\partial Y}{\partial K}=A \alpha Z^{(1/ρ)-1}K^{ρ-1}$$

$$E_K=A \alpha Z^{(1/ρ)-1}K^{ρ-1}*\frac{K}{AZ^{1/ρ}}$$
$$E_K=\alpha \frac{K^ρ}{Z}$$
$$E_K=\alpha \frac{K^ρ}{\alpha K^ρ+(1-\alpha)L^ρ}$$

$$\frac{\partial Y}{\partial L}=A (1-\alpha) Z^{(1/ρ)-1}L^{ρ-1}$$

$$E_L=(1-\alpha)\frac{L^ρ}{Z}$$
$$E_L=(1-\alpha)\frac{L^ρ}{\alpha K^ρ+(1-\alpha)L^ρ}$$

These elasticities tell us how sensitive the level of output is to an increment to capital and labor for a given level of capital stock, labor stock, output, and the parameter alpha.

</details>

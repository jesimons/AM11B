### Sample Problems 3.1: Marginal Functions

While we haven't focused on this in our class, these problems rely on the fact that the word *marginal* means derivative in an econ context.

In particular, if we have:

profit = revenue - costs

Which can be written as:

$\pi(q) = R(q) -C(q)$
where $q$ =

Then, **marginal profit** is:
$$\frac{d\pi}{dq} = MR(q) - MC(q)$$

A key principle: **The firm maximizes profit when marginal revenue equals marginal cost!**

**Problem 1:**

*$C(q)=150+5q+2q^2 , R(q)=60q$*

    a. Find the marginal cost, revenue, and profit when q=30
    b. Is the firm maximizing profit at this quantity?

$MR(q)=R'(q) --> MR(q)=60 --> 60$ dollars

Part B:
No because the difference between marginal cost and marginal revenue is not zero.

$(d\pi)/dq = MR(q)- MC(q) = 60 dollars - 125 dollars = -65 dollars$

**Problem 2:**

The Ferrero factory, specializing in combined Nutella and Kinder snack packs, has a Total Cost (in thousands of dollars) for producing $q$ thousand snack packs given by the function:

$$C(q) = 0.05q^3 - 3q^2 + 50q + 150$$

Find the Marginal Cost function, $MC(q)$, and calculate the marginal cost when the factory produces 30 thousand snack packs ($q=30$).


**Solution:**



<details>

<summary><b>Click to reveal solution</b></summary>

1. Find the Marginal Cost Function, $MC(q)$

The marginal cost is the first derivative of the total cost function, $C(q)$, with respect to quantity $q$:

$$MC(q) = \frac{dC}{dq} = \frac{d}{dq} (0.05q^3 - 3q^2 + 50q + 150)$$

Applying the power rule for differentiation:
$$MC(q) = 3(0.05)q^{3-1} - 2(3)q^{2-1} + 1(50)q^{1-1} + 0$$
$$\mathbf{MC(q) = 0.15q^2 - 6q + 50}$$

2. Calculate Marginal Cost at $q=30$

Substitute $q=30$ into the $MC(q)$ function to find the marginal cost when $30$ thousand snack packs are produced:

$$MC(30) = 0.15(30)^2 - 6(30) + 50$$
$$MC(30) = 0.15(900) - 180 + 50$$
$$MC(30) = 135 - 180 + 50$$
$$MC(30) = \mathbf{5}$$

3. Interpretation


This means that when the Ferrero factory is already producing $30,000$ snack packs, the cost to produce the next $1,000$ snack packs (i.e., the marginal cost) is approximately $\mathbf{\$5}$ thousand, or $\mathbf{\$5,000}$.

**Problem 5:**

A factory’s total cost of producing $q$ toasters is modeled by  
$$C(q) = 80 + 5q + 0.2q^2.$$

1. Find the marginal cost function $MC(q)$.
2. Compute the marginal cost at $q = 20$.
3. Interpret the answer.


**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>


*1. Find the marginal cost function.*

Marginal cost is the derivative of total cost with respect to quantity:  
$MC(q) = \dfrac{dC}{dq}.$

Next differentiate term by term:
- $\dfrac{d}{dq}(80) = 0$
- $\dfrac{d}{dq}(5q) = 5$
- $\dfrac{d}{dq}(0.2q^2) = 0.4q$

The marginal cost function is  
$MC(q) = 5 + 0.4q.$

*2. Evaluate at $q = 20$.*

$MC(20) = 5 + 0.4(20) = 5 + 8 = 13.$

*Step 3: Interpretation.*

At a production level of 20 toasters, the marginal cost is \$13.  
This means producing *one more toaster (the 21st toaster)* costs approximately \$13 in additional cost.



</details>

**Problem 12:**

The widget producing firm sells their widgets for $50 each. The firm faces a total cost function:
                                               
$$C(q)=200+4q+0.5q^2$$

Find the firm's marginal profit function.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

Since the firm receives $50 for each unit sold,

$$R(q)=50q$$

Since profit is equal to revenues minus costs,

$$\pi(q)=R(q)-C(q)$$
$$\pi(q)=50q-(200+4q+0.5q^2)$$
$$\pi(q)=50q-200-4q-0.5q^2$$
$$\pi(q)=-0.5q^2+46q-200$$

The marginal profit function will be equal to the derivative of the profit function.

$$M\pi(q)=2(-0.5q)+46$$
$$M\pi(q)=-q+46$$

This means that for any quantity q of widgets the firm produces, the change in profit received from selling the q-th unit is equal to the resulting function output for that value q.

</details>

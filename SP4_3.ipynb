**Problem 12:**

A wildlife reserve introduces **50** rabbits into a fenced meadow. The meadow can sustainably support at most **500** rabbits. After **4 months**, a population survey finds **150** rabbits.

1. Fit a logistic growth model $(P(t))$ (population vs. months).  
2. Use the model to predict the population after **10 months**.  
3. How long will it take for the population to reach **half the carrying capacity**?

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

We use the logistic model
$
\frac{dP}{dt}=rP\left(1-\frac{P}{K}\right)
$
with solution
$
P(t)=\frac{K}{1 + \left(\dfrac{K-P_0}{P_0}\right)e^{-rt}}.
$

**Step 1 — identify known values**

- $(P_0 = 50)$ (initial population at $(t=0)$)  
- $(K = 500)$ (carrying capacity)  
- $(P(4) = 150)$ (after 4 months)

Compute the constant factor
$
A=\frac{K-P_0}{P_0}=\frac{500-50}{50}=\frac{450}{50}=9.
$

So,
$
P(t)=\frac{500}{1+9e^{-rt}}.
$

Use $(P(4)=150)$ to solve for $(r)$:
$
150=\frac{500}{1+9e^{-4r}}
\quad\Rightarrow\quad
1+9e^{-4r}=\frac{500}{150}=\frac{10}{3}.
$
    
Thus,
$
9e^{-4r}=\frac{10}{3}-1=\frac{7}{3}
\quad\Rightarrow\quad
e^{-4r}=\frac{7}{27}.
$

Take logs:
$
-4r=\ln\!\left(\frac{7}{27}\right)
\quad\Rightarrow\quad
r=\frac{1}{4}\ln\!\left(\frac{27}{7}\right).
$

Numerically,
$
r \approx 0.3375\ \text{per month}.
$

**Final model:**
$
\boxed{\,P(t)=\frac{500}{1+9e^{-0.3375\,t}}\,}
$
(You can keep $(r=\tfrac{1}{4}\ln(27/7))$ for an exact form.)


**Step 2 — predict $(P(10))$**

Compute $(e^{-0.3375\cdot 10}\approx e^{-3.375}\approx 0.0344)$.

Then denominator $(1+9(0.0344)\approx 1.3096)$, so
$
P(10)\approx \frac{500}{1.3096}\approx 382.
$

**Prediction:** After 10 months there will be about **382 rabbits**.


**Step 3 — time to reach half the carrying capacity $(K/2=250)$**

Set $(P(t)=250)$:
$
\frac{500}{1+9e^{-rt}}=250
\quad\Rightarrow\quad
1+9e^{-rt}=2
\quad\Rightarrow\quad
9e^{-rt}=1
$

So,
$
e^{-rt}=\frac{1}{9}\quad\Rightarrow\quad -rt=\ln\!\left(\tfrac{1}{9}\right)=-\ln 9
$
$
t=\frac{\ln 9}{r}.
$

Using $(r\approx0.3375)$:
$
t\approx\frac{2.1972246}{0.3375}\approx 6.51\ \text{months}.
$

**Answer:** It takes about **6.5 months** to reach half the carrying capacity.

- Growth slows as $(P)$ approaches $(K=500)$; long-run equilibrium is $(P(t)\to500)$.  
- The intrinsic rate $(r\approx0.3375)$ (per month) measures how quickly the population responds when small.

**Problem**
**A lake is stocked with fish and the population follows a logistic model with carrying capacity \(K=600\). Initially $ \ P(0)=40\ $
After one year the population is observed to be $ \ P(1)=55\ $**

**Logistic model**
$$ \ \frac{dP}{dt}=rP\left(1-\frac{P}{K}\right) \ $$

**Step A: Find the constant $ \ A\ $ (the integration constant written as a ratio) using $ \ P(0)=40\ $**

**The general logistic solution may be written**
$$ \ P(t)=\frac{K}{1+A e^{-rt}},\qquad A=\frac{K-P_0}{P_0} \ $$

**With** $ \ K=600\ $ and $ \ P_0=40\ $,
$$ \ A=\frac{600-40}{40}=\frac{560}{40}=14 \ $$

**Step B: Use the observation \(P(1)=55\) to find the growth rate \(r\) $**

**Plug$ \ t = 1 \ $ into the solution:**
$$ \ 55 = \frac{600}{1+14e^{-r}} \ $$

**Rearrange:**
$$ \ 1+14e^{-r}=\frac{600}{55}=10.9090909\ldots \ $$
$$ \ 14e^{-r}=9.9090909\ldots\quad\Rightarrow\quad e^{-r}=\frac{9.9090909\ldots}{14}=0.7077922078\ldots\ $$

**Take natural log:**
$$ \ -r=\ln(0.7077922078\ldots)\quad\Rightarrow\quad r\approx -\ln(0.7077922078)\approx 0.3454 \ $$

**So**
$$ \ \boxed{r\approx 0.345\ \text{per year}} \ $$

**Step C:  Write the explicit logistic function $ \ P(t)\ $**

**Using** $$ \ K=600,\ A=14,\ r\approx0.3454 \ $$
$$\ \boxed{\,P(t)=\frac{600}{1+14e^{-0.3454\,t}}\,.} \ $$

**Step D: When does the population reach half the carrying capacity** $$ (i.e. \ P(t)=300 \ )? $$

**Set $ \ P(t)=30 \ $:**
$$ \ 300=\frac{600}{1+14e^{-0.3454\,t}} \quad\Rightarrow\quad 1+14e^{-0.3454\,t}=2 \ $$
$$ \ 14e^{-0.3454\,t}=1 \quad\Rightarrow\quad e^{-0.3454\,t}=\frac{1}{14} \ $$
$$ \ -0.3454\,t=\ln\left(\frac{1}{14}\right)=-\ln 14 \quad\Rightarrow\quad t=\frac{\ln 14}{0.3454}\approx\frac{2.639057}{0.3454}\approx 7.64\ \text{years} \ $$

**So the population reaches 300 fish at approximately $ \ t\approx 7.64\ $ years**

**Step E: When does the population reach 90\% of $ \ K \ $ (i.e. $ \ P(t) = 540 \ $ ?**

**Set $ \ P(t) = 540 \ $ :**
$$ \ 540=\frac{600}{1+14e^{-0.3454\,t}} \quad\Rightarrow\quad 1+14e^{-0.3454\,t}=\frac{600}{540}=1.111111\ldots \ $$
$$ \ 14e^{-0.3454\,t}=0.111111\ldots \quad\Rightarrow\quad e^{-0.3454\,t}=0.007936508\ldots \ $$
$$ \ -0.3454\,t=\ln(0.007936508)\approx -4.8378 \quad\Rightarrow\quad t\approx \frac{4.8378}{0.3454}\approx 14.01\ \text{years} \ $$

**So 90\% of carrying capacity is reached at about $ \  t \approx 14.0 \ $ years**

$$ \ =0.3454\cdot 40\cdot\frac{560}{600} =0.3454\cdot 37.3333\ldots\approx 12.89\ \text{fish per year} \ $$

A fish population in a lake starts at 50 fish. The lake can hold 200 fish, and the population grows at a rate of 0.4 per year.

1. Write the logistic growth model for this situation.
2. Predict the population after 5 years.


**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

We use the logistic growth formula:

$$
P(t) = \frac{K}{1 + \left(\frac{K}{P_0} - 1\right)e^{-rt}}
$$

Where:

- $P_0 = 50$ (initial population)
- $K = 200$ (carrying capacity)
- $r = 0.4$ (growth rate)
- $t$ = time in years


### Step 1: Find the model

Plug in the values:

$$
P(t) = \frac{200}{1 + \left(\frac{200}{50} - 1\right)e^{-0.4t}}
$$

Compute the term inside the parentheses:

$$
\frac{200}{50} = 4 \quad \text{so} \quad 4 - 1 = 3
$$

So the model becomes:

$$
P(t) = \frac{200}{1 + 3e^{-0.4t}}
$$

### Step 2: Predict the population after 5 years

Plug in $t = 5$:

$$
P(5) = \frac{200}{1 + 3e^{-0.4\cdot 5}}
$$

Compute the exponent:

$$
-0.4 \cdot 5 = -2
$$

So:

$$
P(5) = \frac{200}{1 + 3e^{-2}}
$$

We approximate:

$$
e^{-2} \approx 0.135
$$

So:

$$
1 + 3(0.135) = 1 + 0.405 = 1.405
$$

Finally:

$$
P(5) \approx \frac{200}{1.405} \approx 142.
$$


### Final Answer

After 5 years, the population will be about 142 fish

This makes sense because:

- The population is growing.
- But it cannot grow forever.
- It slows down as it approaches 200, the carrying capacity.

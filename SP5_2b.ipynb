**Problem 2:**

Let $f(x,y) = x^3y^2 + 5x - 7y^4$. Find the second-order partial derivatives $f_xx$, $f_yy$, and the mixed partial derivative $f_xy$. Finally, verify that $f_xy$ = $f_yx$ (Clairaut's Theorem).

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

Step 1: Compute the First-Order Partial Deriviates

- $f_x$ (Differentiate with respect to $x$): $3x^2y^2 + 5$
- $f_y$ (Differentiate with respect to $y$): $2x^3y - 28y^3$

Step 2: Compute the Second-Order Partial Derivatives

- Find $f_xx$: Differentiate $f_x$ with respect to $x$ again.

    $$f_{xx} = \frac{\partial}{\partial x}(3x^2y^2 + 5) = 6xy^2$$

- Find $f_yy$: Differentiate $f_y$ with respect to $y$ again.

    $$f_{yy} = \frac{\partial}{\partial y}(2x^3y - 28y^3) = 2x^3 - 84y^2$$

- Find $f_xy$: Differentiate $f_x$ with respect to $y$.

    $$f_{xy} = \frac{\partial}{\partial y}(3x^2y^2 + 5) = 6x^2y$$

Step 3: Verify Clairaut's Theorem. The theorem states that for continuous functions, the mixed partials are equal ($f_xy$ = $f_yx$).

- We found $f_xy$ = $6x^2y$
- Now find $f_yx$ (Take $f_y$ and differentiate with respect to $x$):

    $$f_{yx} = \frac{\partial}{\partial x}(2x^3y - 28y^3) = 6x^2y$$

Conclusion: Since $6x^2y = 6x^2y$, we've verified the theorem.

</details>

**Problem 20:**

*Let*
$$
f(x,y) = x^3 y^2 + 4x^2 y - 5y^2.
$$
*Compute the second-order partial derivatives $f_{xx}$, $f_{yy}$, and $f_{xy}$. Then verify that $f_{xy} = f_{yx}$. *

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1: First-order partials**

Treat $y$ as a constant for $f_x$:
$$
f(x,y) = x^3 y^2 + 4x^2 y - 5y^2.
$$

Differentiate term-by-term with respect to $x$:

- $\dfrac{\partial}{\partial x}(x^3 y^2) = 3x^2 y^2$  
- $\dfrac{\partial}{\partial x}(4x^2 y) = 8xy$  
- $\dfrac{\partial}{\partial x}(-5y^2) = 0$

So
$$
f_x(x,y) = 3x^2 y^2 + 8xy.
$$

Now treat $x$ as a constant for $f_y$:
$$
f_y(x,y) = \dfrac{\partial}{\partial y}(x^3 y^2)
          + \dfrac{\partial}{\partial y}(4x^2 y)
          + \dfrac{\partial}{\partial y}(-5y^2)
        = 2x^3 y + 4x^2 - 10y.
$$

**Step 2: Second-order partials**

Now differentiate $f_x$ again with respect to $x$ to get $f_{xx}$:
$$
f_{xx}(x,y) = \dfrac{\partial}{\partial x}\big(3x^2 y^2 + 8xy\big)
            = 6x y^2 + 8y.
$$

Differentiate $f_y$ again with respect to $y$ to get $f_{yy}$:
$$
f_{yy}(x,y) = \dfrac{\partial}{\partial y}\big(2x^3 y + 4x^2 - 10y\big)
            = 2x^3 - 10.
$$

**Step 3: Mixed partials and Clairaut’s theorem**

First do $f_{xy}$ by differentiating $f_x$ with respect to $y$:
$$
f_{xy}(x,y) = \dfrac{\partial}{\partial y}\big(3x^2 y^2 + 8xy\big)
            = 6x^2 y + 8x.
$$

Now do $f_{yx}$ by differentiating $f_y$ with respect to $x$:
$$
f_{yx}(x,y) = \dfrac{\partial}{\partial x}\big(2x^3 y + 4x^2 - 10y\big)
            = 6x^2 y + 8x.
$$

verified that $f_{xy} = f_{yx}$, illustrating Clairaut’s theorem for this function.

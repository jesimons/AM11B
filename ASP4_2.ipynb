**Problem 2:**
 A UC Santa Cruz student derives utility ($U$) from consuming two goods: tote bags ($x$) and matcha lattes ($y$). The student's utility function is given by:
$$U(x, y) = 5x^{0.5}y^{0.5}$$
Derive the Marginal Utility functions for tote bags ($MU_x$) and matcha lattes ($MU_y$).
Derive the formula for the Marginal Rate of Substitution ($\text{MRS}$).
Calculate the $\text{MRS}$ when the student currently has $x = 4$ tote bags and $y = 1$ matcha latte.
Interpret the economic meaning of the calculated $\text{MRS}$.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>
1. Calculate Marginal Utilities}

The utility function is $U(x, y) = 5x^{0.5}y^{0.5}$, where $x$ is tote bags and $y$ is matcha lattes.

Marginal Utility of Tote Bags ($MU_x$)
$$MU_x = \frac{\partial U}{\partial x} = \frac{\partial}{\partial x}(5x^{0.5}y^{0.5}) = \mathbf{2.5 y^{0.5} x^{-0.5}}$$

Marginal Utility of Matcha Lattes ($MU_y$)
$$MU_y = \frac{\partial U}{\partial y} = \frac{\partial}{\partial y}(5x^{0.5}y^{0.5}) = \mathbf{2.5 x^{0.5} y^{-0.5}}$$

2. Derive the MRS Formula

The Marginal Rate of Substitution ($\text{MRS}$) is the ratio of marginal utilities:
$$\text{MRS} = \frac{MU_y}{MU_x} = \frac{2.5 x^{0.5} y^{-0.5}}{2.5 y^{0.5} x^{-0.5}}$$
$$\mathbf{\text{MRS} = \frac{x}{y}}$$

3. Calculate $\text{MRS}$ at $(x=4, y=1)$

Substitute $x=4$ (tote bags) and $y=1$ (matcha latte) into the $\text{MRS}$ formula:
$$\text{MRS}(4, 1) = \frac{4}{1} = \mathbf{4}$$

4. Economic Interpretation

The result $\mathbf{\text{MRS} = 4}$ means that at this consumption level, the student is willing to give up approximately $\mathbf{4}$ units of the good on the vertical axis (matcha lattes, $y$) to gain $\mathbf{1}$ additional unit of the good on the horizontal axis (tote bags, $x$), while maintaining the same level of utility.


**Problem:**

*Max is preparing for a week of hiking trips, and needs to pack two types of snacks: smoothies (x) and granola bars (y). His satisaction with each snack is described by the utility function $U(x,y) = x^{0.5}y^{0.5}$. Currently, he is considering packing 4 smoothies and 9 granola bars to conserve space.*

*a) Find Max’s marginal utility of smoothies $MU_x$ and marginal utility of granola bars $MU_y$.*

*b) Compute his MRS and evaluate at $(4,9)$.*

*c) Interpret economically: What does his MRS tell you about how much Max values smoothies relative to granola bars at this moment?*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Find $MU_x$ and $MU_y$.**

*$MU_x = \frac{\partial U}{\partial x} = 0.5x^{-0.5}y^{0.5}$*

*$MU_y = \frac{\partial U}{\partial y} = 0.5x^{0.5}y^{-0.5}$*

**Part (b): Compute MRS**

*$MRS = \frac{MU_y}{MU_x} = \frac{0.5x^{0.5}y^{-0.5}}{0.5x^{-0.5}y^{0.5}} = \frac{x^{0.5}}{x^{-0.5}}⋅\frac{y^{-0.5}}{y^{0.5}} = x^{(0.5+0.5)} ⋅ y^{(-0.5-0.5)} = x^{1}y^{-1} = \frac{x}{y}$*

*$MRS = \frac{4}{9}$*

**Part (c): Economic Interpretation**

*Max's $MRS = \frac{4}{9}$, meaning that he is willing to give up $\frac{4}{9}$ of a smoothie to gain 1 additional granola bar and remain equally satisfied. This means that granola bars are slightly more valuable to him, since he only has to give up a small part of a smoothie to get another granola bar.*

</details>

A consumer gets utility from two goods: cans of soda $x$ and pizza slices $y$.  
Their utility function is

$$U(x,y) = \sqrt{x}\,\sqrt{y} = x^{0.5} y^{0.5}.$$

1. Find the marginal utility of soda, $MU_x = \dfrac{\partial U}{\partial x}$, and the marginal utility of pizza, $MU_y = \dfrac{\partial U}{\partial y}$.
2. Find the marginal rate of substitution (MRS) between soda and pizza:  
   $$MRS = \frac{MU_y}{MU_x}.$$
3. Evaluate the MRS at the bundle $(x,y) = (4,2)$.
4. Interpret what this MRS means in words.

*Step 1: Compute $MU_x$ and $MU_y$.*

Given
$$U(x,y) = x^{0.5} y^{0.5},$$

treat $y$ as constant and differentiate with respect to $x$:
$$
MU_x = \frac{\partial U}{\partial x}
= 0.5 x^{-0.5} y^{0.5}.
$$

Treat $x$ as constant and differentiate with respect to $y$:
$$
MU_y = \frac{\partial U}{\partial y}
= 0.5 x^{0.5} y^{-0.5}.
$$

*Step 2: Find the MRS.*

By definition,
$$
MRS = \frac{MU_y}{MU_x}
= \frac{0.5 x^{0.5} y^{-0.5}}{0.5 x^{-0.5} y^{0.5}}
= \frac{x}{y}.
$$

So
$$
MRS(x,y) = \frac{x}{y}.
$$

*Step 3: Evaluate at $(x,y) = (4,2)$.*

$$
MRS(4,2) = \frac{4}{2} = 2.
$$

*Step 4: Interpretation.*

At the bundle $(x,y) = (4,2)$, the MRS is $2$.  
This means the consumer is willing to give up 2 cans of soda to get 1 more slice of pizza while remaining equally satisfied (on the same indifference curve).

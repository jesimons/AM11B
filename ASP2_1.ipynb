### Returns to Scale

This concept is about the relationship between changes in inputs versus outputs.

In particular, if we double all inputs, what happens to output? Here are the possibilities:

- **Constant returns to scale**: Output exactly doubles
- **Increasing returns to scale**: Output more than doubles (economies of scale)
- **Decreasing returns to scale**: Output less than doubles

**Problem 5:**

*[Create a problem with a specific production function. You may use a Cobb-Douglas equation or another production function form, but specify the constants.]*

Potential questions include:

Suppose a firm's technology is described by the Cobb-Douglas production function:
$$Q(L, K) = 3L^{0.7}K^{0.5}$$
where $Q$ is output, $L$ is labor input, and $K$ is capital input. The constants are $A=3$ (technology constant), $\alpha=0.7$ (labor exponent), and $\beta=0.5$ (capital exponent).

*a) Given input $L$ = $8$ and $K$ = $16$, how much output is produced?*

*b) Does this production function have constant, increasing, or decreasing returns to scale?*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Calculate Output**

Substitute the given values $L=8$ and $K=16$ into the production function:
$$Q(8, 16) = 3(8)^{0.7}(16)^{0.5}$$

Step 1: Calculate $K^{0.5}$:
$$K^{0.5} = 16^{0.5} = \sqrt{16} = 4$$

Step 2: Calculate $L^{0.7}$:
Since $8 = 2^3$, we have:
$$8^{0.7} = (2^3)^{0.7} = 2^{2.1} \approx 4.287$$

Step 3: Calculate the total output:
\begin{align*}
Q(8, 16) &= 3(8^{0.7})(4) \\
&= 12(8^{0.7}) \\
&\approx 12(4.287) \\
&\approx 51.444
\end{align*}

$$Q(8, 16) \approx 51.444$$

**Part (b): Returns to Scale**

Returns to scale for a Cobb-Douglas function $Q(L, K) = AL^\alpha K^\beta$ are determined by the sum of the exponents $\alpha + \beta$. Let $S = \alpha + \beta$.

Step 1: Calculate the sum of the exponents:
$$S = \alpha + \beta = 0.7 + 0.5 = 1.2$$

Step 2: Compare $S$ to 1:
Since $S = 1.2 > 1$, the production function exhibits Increasing Returns to Scale (IRS).

We multiply both inputs by a scaling factor $t > 1$:
\begin{align*}
Q(tL, tK) &= 3(tL)^{0.7}(tK)^{0.5} \\
&= 3t^{0.7}L^{0.7}t^{0.5}K^{0.5} \\
&= t^{0.7+0.5} \cdot [3L^{0.7}K^{0.5}] \\
&= t^{1.2} \cdot Q(L, K)
\end{align*}
Since $1.2 > 1$, we have $t^{1.2} > t$ for any $t>1$.
Therefore, $Q(tL, tK) > t \cdot Q(L, K)$.

The production function has Increasing Returns to Scale (IRS). Doubling the inputs (i.e., $t=2$) more than doubles the output ($2^{1.2} \approx 2.296$ times the original output).

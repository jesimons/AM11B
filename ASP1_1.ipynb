Demand, D(q)= 100-2q
Supply, S(q)= 20+2q

*a) Find the equilibrium price and quantity.*
*b) Calculate the consumer surplus.*
*c) Calculate the producer surplus.*
*d) What is the total economic surplus?*
**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Finding Equilibrium**

Demand: $D(q)=100-2q$  
Supply: $S(q)=20+2q$

To find the equilibrium, set demand equal to supply, $D(q) = S(q)$:
\begin{align*}
100 - 2q &= 20 + 2q \\
100 - 20 &= 2q + 2q \\
80 &= 4q \\
q^* &= \frac{80}{4} \\
q^* &= 20
\end{align*}

Substitute the equilibrium quantity $q^* = 20$ into either the supply or demand function to find the equilibrium price $p^*$:
$$p^* = D(20) = 100 - 2(20) = 100 - 40 = 60$$
Result:
The equilibrium quantity is $q^* = 20$ and the equilibrium price is $p^* = 60$.
$$(q^*, p^*) = (20, 60)$$

**Part (b): Consumer Surplus**
    $$CS = \int_0^{q^*} [D(q) - p^*] \, dq$$
Step 1: Substitute the specific functions:
\begin{align*}
CS &= \int_{0}^{20} \big[(100 - 2q) - 60\big]\, dq \\
   &= \int_{0}^{20} (40 - 2q)\, dq
\end{align*}

Step 2: Compute the antiderivative:
\begin{align*}
\int (40 - 2q)\, dq &= 40q - q^2
\end{align*}

Step 3: Evaluate the integral from 0 to 20:
\begin{align*}
CS &= \left[ 40q - q^2 \right]_0^{20} \\
   &= (40 \cdot 20 - 20^2) - (0) \\
   &= 800 - 400 \\
   &= 400
\end{align*}

Result:
CS = 400

Interpretation:The value of $\text{CS}=400$ represents the net economic benefit or gain in utility that consumers receive from purchasing the good at the equilibrium price ($P^*=60$).

**Part (c): Producer Surplus**

$$PS = \int_0^{q^*} [p^* - S(q)] \, dq$$

Step 1: Substitute the specific functions and equilibrium values $(q^*=20, p^*=60)$:
\begin{align*}
PS &= \int_{0}^{20} \big[60 - (20 + 2q)\big]\, dq \\
   &= \int_{0}^{20} (60 - 20 - 2q)\, dq \\
   &= \int_{0}^{20} (40 - 2q)\, dq
\end{align*}

Step 2: Compute the antiderivative:
\begin{align*}
\int (40 - 2q)\, dq &= 40q - q^2
\end{align*}

Step 3: Evaluate the integral from 0 to 20:
\begin{align*}
PS &= \left[ 40q - q^2 \right]_0^{20} \\
   &= (40 \cdot 20 - 20^2) - (0) \\
   &= 800 - 400 \\
   &= 400
\end{align*}

Result:
$$PS = 400$$

Interpretation: The producer surplus is 400. This is the total benefit producers receive from selling the product in the market, measured as the difference between the market price they receive and the minimum price they would have been willing to accept (the supply curve).


**Part (d): Total Surplus**

*[Total Surplus = CS + PS]*

Step 1: Substitute the calculated values for Consumer Surplus $(CS=400)$ and Producer Surplus $(PS=400)$:
\begin{align*}
TS &= 400 + 400 \\
   &= 800
\end{align*}


$$TS = 800$$

Interpretation: The Total Economic Surplus is 800. This value represents the total net benefit to society (both consumers and producers) from the production and consumption of this good at the equilibrium price and quantity. Since the total surplus is maximized at the competitive equilibrium, this represents the highest possible welfare for this market.

*A local artisan market introduces a new handmade product: solar lanterns, small rechargeable lanterns that absorb sunlight during the day and glow at night. They quickly become popular among hikers and students who enjoy outdoor study spaces. Market researchers estimate that the weekly demand for the lanters is: $P = 48 - 2q$, where P is the price in dollars and q is the number of lanterns demanded per week. However, local artisans have limited capacity and higher production costs. Their weekly supply curve is: $P = 12 + q$.*

*a) Find the equilibrium price and equilibrium quantity of the solar lanterns.*

*b) Calculate the consumer surplus at equilibrium.*

*c) Calculate the producer surplus at equilibrium.*

*d) What is the total economic surplus created in this market?*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Finding Equilibrium**

*Set D(q) = S(q) and solve for q:*

*$48 - 2q = 12 + q$*

*$48 - 12 = 2q + q$*

*$36 = 3q$*
    
*$q = \frac{36}{3} = 12$*

*Find the equilibrium price:*

*$P = 12 + q = 12 + 12 = 24$ dollars*

*OR*

*$P = 48 - 2q = 48 - 2(12) = 48 - 24 = 24$ dollars*

**Part (b): Consumer Surplus**

$$CS = \int_0^{q^*} [D(q) - P] \, dq$$

*Input q, D(q) and P into the integral:*

*$CS = \int_0^{12} [(48 - 2q) - 24] \, dq$*

*Simplify the inside:*

*$48 - 24 = 24$*

*So, $CS = \int_0^{12} [24 - 2q] \, dq$*

*Integrate step-by-step:*

*$CS = \int_0^{12} [24 - 2q] \, dq = \int 24 \, dq - \int 2q \, dq$*

*$\int 24 \, dq = 24q$*

*$\int 2q \, dq = q^2$*

*$CS = \int_0^{12} [24 - 2q] \, dq = 24q - q^2$*
    
*Evaluate from 0 to 12:*

*$CS = [24q - q^2]_0^{12} = [24(12) - 12^2] - [24(0) - (0)^2] = [288 - 144] - 0 = 144$*

*The Consumer Surplus is 144 dollars, meaning that buyers gain 144 dollars in extra satisfaction/value each week from buying solar lanterns at the market price of 24 dollars.*

**Part (c): Producer Surplus**

$$PS = \int_0^{q^*} [P - S(q)] \, dq$$

*Input q, S(q) and P into the integral:*

*$PS = \int_0^{12} [24 - (12 + q)] \, dq$*

*Simplify the inside:*

*$24 - (12 + q) = 24 - 12 - q = 12 - q$*

*So, $CS = \int_0^{12} [12 - q] \, dq$*

*Integrate step-by-step:*

*$CS = \int_0^{12} [12 - q] \, dq = \int_0^{12} 12 \, dq - \int_0^{12} q \, dq$*

*$\int 12 \, dq = 12q$*

*$\int q \, dq = \frac{q^2}{2}$*

*$CS = \int_0^{12} [12 - q] \, dq = 12q - \frac{q^2}{2}$*

*Evaluate from 0 to 12:*

*$CS = [12q - \frac{q^2}{2}]_0^{12} = [12(12) - \frac{12^2}{2}] - [12(0) - \frac{0^2}{2}(0)] = [144 - 72] - 0 = 72$*

*The Producer Surplus is 72 dollars, meaning that artisans benefit 72 extra dollars per week compared to the minimum they would be willing to accept.*

**Part (d): Total Surplus**

*[Total Surplus = CS + PS]*

*CS = $144*

*PS = $72*

*Total Surplus = 144 + 72 = $216 dollars*

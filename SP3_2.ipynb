**Problem 1:**
$\int$ $\frac {2x + 3}{x^2 - x - 6}dx$

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1)** Factor denominator.

$ x^2 - x - 6 = (x - 3)(x + 2)$

**Step 2)** Set up the partial fraction form.

$\frac {2x + 3}{x^2 - x - 6}$ = $\frac {A}{x-3} + \frac {B}{x+2}$

**Step 3)** Multiply both sides by $(x - 3)(x + 2)$.

$= 2x + 3 = A(x + 2) + B(x - 3)$

**Step 4)** Solve for A and B.

Rewrite: $2x + 3 = (A + B)x + (2A - 3B)$

$A + B = 2$    and   $2A - 3B = 3$

Solve for B:

$ A= 2 - B$

Plug in A:

$2 (2 - B) - 3B$ = $4 - 2B - 3B$ = $4 - 5B$ = $3$

$-5B = -1$

$B = \frac {1}{5}$

Plug in B to solve for A: $A = 2 - B$ = $2 - \frac {1}{5}$ = $\frac {9}{5}$

**Step 5)** Rewrite the integral with the solutions of A and B with respect to x.

$= \frac {\frac {9}{5}}{x-3}$ + $\frac {\frac{1}{5}}{x+2}$

**Step 6)** Integrate all terms, and that will be your solution.

$\int$ $\frac {\frac {9}{5}}{x-3}$ + $\frac {\frac{1}{5}}{x+2}$ $dx$ =  $\frac {9}{5}$ $\int \frac {1}{x-3} dx$ + $\frac {1}{5}$ $\int \frac {1}{x+2} dx$

$= \frac {9}{5}ln|x-3|$ + $\frac {1}{5} ln |x+2| + C$

**Problem 11:**  
Decompose the following rational function into partial fractions and integrate:

$$\int \frac{3x + 5}{(x-1)(x+2)} \, dx $$

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1: Set up the partial fraction decomposition**

We assume:

$$ \frac{3x + 5}{(x-1)(x+2)} = \frac{A}{x-1} + \frac{B}{x+2} $$
where \(A\) and \(B\) are constants to be determined.

**Step 2: Multiply both sides by the denominator \((x-1)(x+2)\)**

$$ 3x + 5 = A(x + 2) + B(x - 1) $$

**Step 3: Expand the right-hand side**

$$ 3x + 5 = A x + 2A + B x - B = (A + B)x + (2A - B) $$

**Step 4: Solve for \(A\) and \(B\) by equating coefficients**

For the (x)-term: $$ (A + B = 3) $$  
For the constant term: $$ (2A - B = 5) $$

Solve the system:

1. $$ (A + B = 3) $$  
2. $$ (2A - B = 5) $$

Add the two equations: $$ ( (A + B) + (2A - B) = 3 + 5 ) $$ → $$ ( 3A = 8 ) $$ → $$ ( A = \frac{8}{3} ) $$  

Then $$ ( B = 3 - A = 3 - \frac{8}{3} = -\frac{5}{3} ) $$

**Step 5: Rewrite the integrand**

$$ \frac{3x + 5}{(x-1)(x+2)} = \frac{8/3}{x-1} - \frac{5/3}{x+2} $$

**Step 6: Integrate term by term**

$$ \int \frac{3x + 5}{(x-1)(x+2)} dx = \frac{8}{3} \int \frac{1}{x-1} dx - \frac{5}{3} \int \frac{1}{x+2} dx $$

$$ = \frac{8}{3} \ln|x-1| - \frac{5}{3} \ln|x+2| + C $$

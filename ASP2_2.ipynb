### Utility Functions: Background Info

A **utility function** represents consumer preferences over different "bundles" of goods. If we have quantities of two goods (call these quantities x and y), specific values of (x,y) are called a bundle. Then, a utility function is a function of two variables:
$$U(x, y)$$

### Indifference Curves
 An **indifference curve** connects all bundles (x,y) that give the **same utility level**. The consumer is indifferent between any two points on the same curve. In the plot above, an indifference curve can be visualized as any one of the borders between two different colors.

 To find indifference curves, we set $U$ as a constant, called $\bar{U}$. Then, we can solve for $x_2$ as a function of $x_1$ from the equation:
$$\bar{U} = U(x_1, x_2)$$
$$x_2 = \cdots$$
This $x_2$ is what we will call an indifference curve. (It's just a relationship between $x_1$ and $x_2$.

**Problem 2:**

*A consumer named Mia buys two goods: coffee ($x_1$) and pastries ($x_2$). Her satisfaction (utility) from consuming these goods is modeled by the Cobb–Douglas utility function: $U(x_1, x_2) = 3x_1^{0.4}x_2^{0.6}$. The values 0.4 and 0.6 represent how strongly he values pastries over coffee.*

*a) Mia is considering between 4 different breakfast bundles on the way to work one morning. Compute the utility for each one and determine which has the most value to Mia:*

- Bundle 1: 1 cup of coffee and 1 pastry

- Bundle 2: 2 cups of coffee and 1 pastry

- Bundle 3: 1 cup of coffee and 4 pastries

- Bundle 4: 5 cups of coffee and 2 pastries

*b) Derive the general indifference curve function.*
    
*c) Solve for each indifference curve set the utility levels.*




**Solution:**


<details>
<summary><b>Click to reveal solution</b></summary>

**Part (a): Compute utility for each bundle using the formula $U(x_1, x_2) = 3x_1^{0.4}x_2^{0.6}$.**

*Bundle 1: (1, 1)*

*$U(1,1) = 3(1)^{0.4}(1)^{0.6} = 3(1)(1) = 3$*

*Bundle 2: (2, 1)*

*$U(2,1) = 3(2)^{0.4}(2)^{0.6} ≈ 3(1.320)(1) ≈ 3.959$*

*Bundle 3: (1, 4)*

*$U(1,4) = 3(1)^{0.4}(4)^{0.6} ≈ 3(1)(2.296) = 6.888$*

*Bundle 4: (5, 2)*

*$U(5,2) = 3(5)^{0.4}(2)^{0.6} ≈ 3(1.904)(1.516) ≈ 8.653$*

Bundle 4 (5 coffees, 2 pastries) has the highest value to Mia because it gives the highest utility value (8.653).

**Part (b): Derive the general indifference curve function.**

*Solve for $x_2$ using $U(x_1, x_2) = 3x_1^{0.4}x_2^{0.6}$:*

*Divide both sides by 3.*

*$x_1^{0.4}x_2^{0.6} = \frac{U}{3}$*

*Isolate $x_2^{0.6}$.*

*$x_2^{0.6} = \frac{U_0}{3}(x_1^{-0.4})$*

*Raise both sides to power $\frac{1}{0.6}$ (aka $\frac{5}{3}$).*

*$x_2 = (\frac{U_0}{3})^{\frac{1}{0.6}}x_1^{\frac{-0.4}{0.6}} = (\frac{U_0}{3})^{\frac{5}{3}}x_1^{\frac{-2}{3}}$*

*$x_2 = (\frac{U_0}{3})^{\frac{5}{3}}x_1^{\frac{-2}{3}}$*

**Part (c): Solve for each indifference curve set for each utility level and identify the constant $k$.**

*Plug in each $U_0$ from Part a) into the equation $x_2 = kx_1^{\frac{-2}{3}}$, where $k = (\frac{U_0}{3})^{\frac{5}{3}}$.*

*For Bundle 1 $(U_0 = 3)$:*

*$x_2 = (\frac{3}{3})^\frac{5}{3} = 1$*

*$k = 1$*

*$x_2 = (1)x_1^{\frac{-2}{3}} = x_1^{\frac{-2}{3}}$*

*For Bundle 2 $(U_0 = 3(2)^{0.4})$:*

*$x_2 = (\frac{3(2)^{0.4}}{3})^\frac{5}{3} = (2^{0.4})^\frac{5}{3} = 2^\frac{0.4(5)}{3} = 2^\frac{2}{3}$*

*$k = 2^\frac{2}{3}$*

*$x_2 = 2^\frac{2}{3}x_1^{\frac{-2}{3}}$*

*For Bundle 3 $(U_0 = 3(4)^{0.6})$:*

*$x_2 = (\frac{3(4)^{0.6}}{3})^\frac{5}{3} = (4^{0.6})^\frac{5}{3} = 4^\frac{0.6(5)}{3} = 4^1 = 4$*

*$k = 4$*

*$x_2 = 4x_1^{\frac{-2}{3}}$*

*For Bundle 4 $(U_0 = 3(5)^{0.4}(2)^{0.6})$:*

*$x_2 = (\frac{3(5)^{0.4}(2)^{0.6}}{3})^\frac{5}{3} = (5^{0.4}2^{0.6})^\frac{5}{3} = 5^\frac{0.4(5)}{3}2^\frac{0.6(5)}{3} = 5^\frac{2}{3}2^1 = 2(5)^\frac{2}{3}$*

*$k = 2(5)^\frac{2}{3}$*

*$x_2 = 2(5)^\frac{2}{3}x_1^{\frac{-2}{3}} ≈ 5.848x_1^{\frac{-2}{3}}$*

*Interpretation: All of the indifference curves have have the form $x_2 = kx_1^{\frac{-2}{3}}$ with different constants $k$, and are convex with a downward slope. Larger $k$ value corresponds to a higher utility value (higher satisfaction/value in a product).*

**Problem 5:**
  
A consumer views peanut butter ($x_1$) and jelly ($x_2$) as complementary goods. Their preferences are represented by the Cobb-Douglas utility function:
$$U(x_1, x_2) = 3x_1^{\frac{1}{2}}x_2^{\frac{1}{2}}$$

where the preference parameter is $\alpha = \frac{1}{2}$.

a) Compute $U(1, 1)$

b) Compute $U(2, 5)$

c) Compute $U(9, 3)$

<details>
<summary><b>Click to reveal solution</b></summary>

The utility function is $U(x_1, x_2) = 3\sqrt{x_1 x_2}$.

Part (a): Compute $U(1, 1)$


Substitute $x_1=1$ and $x_2=1$ into the utility function:
\begin{align*}
U(1, 1) &= 3(1)^{\frac{1}{2}}(1)^{\frac{1}{2}} \\
&= 3 \cdot 1 \cdot 1 \\
&= 3
\end{align*}

Result:
$$U(1, 1) = 3$$

Part (b): Compute $U(2, 5)$


Substitute $x_1=2$ and $x_2=5$ into the utility function:
\begin{align*}
U(2, 5) &= 3(2)^{\frac{1}{2}}(5)^{\frac{1}{2}} \\
&= 3\sqrt{2 \cdot 5} \\
&= 3\sqrt{10}
\end{align*}
Using $\sqrt{10} \approx 3.162$:
$$U(2, 5) \approx 3(3.162) \approx 9.487$$

Result:
$$U(2, 5) = 3\sqrt{10} \approx 9.49$$

Part (c): Compute $U(9, 3)$


Substitute $x_1=9$ and $x_2=3$ into the utility function:
\begin{align*}
U(9, 3) &= 3(9)^{\frac{1}{2}}(3)^{\frac{1}{2}} \\
&= 3\sqrt{9} \cdot \sqrt{3} \\
&= 3 \cdot 3 \cdot \sqrt{3} \\
&= 9\sqrt{3}
\end{align*}
Using $\sqrt{3} \approx 1.732$:
$$U(9, 3) \approx 9(1.732) \approx 15.588$$

Result:
$$U(9, 3) = 9\sqrt{3} \approx 15.59$$

Interpretation: Bundle $(9, 3)$ yields the highest utility ($15.59$), meaning the consumer prefers having more peanut butter (9 units) and jelly (3 units) than the other two bundles. Since $\alpha=1/2$, the consumer has an equal preference weight for both goods.

Consider a consumer who chooses between quantities of good 1 ($x_1$) and good 2 ($x_2$).

1. Compute the utility for the following three bundles:  
A: $(x_1,x_2) = (1,9)$  
B: $(x_1,x_2) = (4,4)$  
C: $(x_1,x_2) = (9,1)$  

   Which bundle gives the highest utility?

2. Find the equation for the indifference curve corresponding to a fixed utility level $\bar U$.
   Write $x_2$ as a function of $x_1$.

For the solution:


**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>


1. Compute utilities:

   - A:
     $$U(1,9) = 1^{1/2} * t 9^{1/2} = 1 * 3 = 3.$$

   - B:
     $$U(4,4) = 4^{1/2} * 4^{1/2} = 2 * 2 = 4.$$

   - C:

   $$U(4,4) = 4 > U(1,9) = 3 = U(9,1) = 3.$$

   $$x_2(x_1) = \frac{\bar U^2}{x_1}.$$

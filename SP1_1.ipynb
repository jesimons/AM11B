**Problem 11:**

*Find the antiderivative of* $$ f(x) = 6x^3 - 2x + 4 $$.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

*Work through the problem step by step. Show all work clearly.*


**Step 1: Split the integral using linearity**
$$ \int (6x^3 - 2x + 4)\,dx
= \int 6x^3\,dx - \int 2x\,dx + \int 4\,dx $$

**Step 2: Pull out constants**
$$ = 6\int x^3\,dx \;-\; 2\int x\,dx \;+\; 4\int 1\,dx $$

**Step 3: Apply the power rule**

The power rule is:
$$ \int x^n\,dx = \frac{x^{n+1}}{n+1} + C \qquad (n \neq -1) $$

Apply it term-by-term:

**For**
$$ ( \int x^3\,dx \)$$ :
$$ \int x^3\,dx = \frac{x^4}{4} $$
$$ 6\int x^3\,dx = 6\cdot\frac{x^4}{4} = \frac{3}{2}x^4. $$

**For** $$ ( \int x\,dx ) $$ (here \(x = x^1\)):
$$ \int x\,dx = \frac{x^2}{2} $$
$$ -2\int x\,dx = -2\cdot\frac{x^2}{2} = -x^2. $$

**For** $$ ( \int 1\,dx \ ) $$ (constant term):
$$ \int 1\,dx = x $$
$$ 4\int 1\,dx = 4x. $$

**Combine the results**
$$ \int (6x^3 - 2x + 4)\,dx
= \frac{3}{2}x^4 - x^2 + 4x + C. $$

**Final Answer**
$$ F(x) = \frac{3}{2}x^4 - x^2 + 4x + C $$

**Check (differentiate to verify)**
$$ F'(x) = \frac{d}{dx}\left(\frac{3}{2}x^4 - x^2 + 4x + C\right)
= 6x^3 - 2x + 4, $$
which matches the original \( f(x) \).

</details>

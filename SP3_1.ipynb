$$ \int x \ln(x)\,dx = \frac{x^2}{2} \ln(x) - \frac{x^2}{4} + C $$

**Problem 23:**

Find $\displaystyle \int x e^x\,dx$.

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1: Choose $u$ and $dv$.**

We have a product $x e^x$.

By LIATE, Algebraic ($x$) comes before Exponential ($e^x$), so we choose  
$u = x$ and $dv = e^x\,dx$.

**Step 2: Find $du$ and $v$.**

Differentiate $u$:

$u = x \quad \Rightarrow \quad du = 1\,dx = dx.$

Integrate $dv$:

$dv = e^x\,dx \quad \Rightarrow \quad v = e^x.$

**Step 3: Apply the integration by parts formula.**

$\displaystyle \int u\,dv = uv - \int v\,du.$

So

$\displaystyle \int x e^x\,dx
= x e^x - \int e^x\,dx.$

**Step 4: Integrate the remaining integral.**

$\displaystyle \int e^x\,dx = e^x,$

so

$\displaystyle \int x e^x\,dx
= x e^x - e^x + C.$

We can factor $e^x$ if we want:

$\displaystyle \int x e^x\,dx
= e^x(x - 1) + C.$

So the final answer is

$\boxed{e^x(x - 1) + C}.$

</details>

**Problem 21:**

*Let the function be*
$$
f(x,y) = x^2y + 3xy^2 - 4y.
$$
*Find the partial derivatives $f_x(x,y)$ and $f_y(x,y)$.*

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Finding \(f_x\):**

*Treat \(y\) as a constant and differentiate with respect to \(x\).*

We have
$$
f(x,y) = x^2y + 3xy^2 - 4y.
$$

Differentiate term-by-term with respect to \(x\):

- $\dfrac{\partial}{\partial x}(x^2y) = 2xy$  (power rule, \(y\) is constant),
- $\dfrac{\partial}{\partial x}(3xy^2) = 3y^2$  (since \(3y^2\) is a constant multiple of \(x\)),
- $\dfrac{\partial}{\partial x}(-4y) = 0$  (no \(x\) in this term).

So
$$
f_x(x,y) = 2xy + 3y^2.
$$

**Finding \(f_y\):**

*Now treat \(x\) as a constant and differentiate with respect to \(y\).*

Again start from
$$
f(x,y) = x^2y + 3xy^2 - 4y.
$$

Differentiate term-by-term with respect to \(y\):

- $\dfrac{\partial}{\partial y}(x^2y) = x^2$,
- $\dfrac{\partial}{\partial y}(3xy^2) = 6xy$,
- $\dfrac{\partial}{\partial y}(-4y) = -4$.

so
$$
f_y(x,y) = x^2 + 6xy - 4.
$$

</details>

**What is a partial derivative?**

- A partial derivative measures how **a multivariable function changes** with respect to **one variable at a time**, while holding all other variables constant.

- For a function of $f(x,y)$, the partial derivative with $x$ in mind, is notated as $\frac{\partial f}{\partial x}$ meaning $f_x$

- The original notation for determening derivatives is $dy/dx$, meaning that $y$ is the dependent variable and $x$ is the independent variable. Partial derivatives take the original notation and replaces $d$ with the symbol ${\partial }$

    
**Notation:**

- $\frac{\partial f}{\partial x}$ or $f_x$: The Partial of $f$ with respect to $x$, written as  $\frac{\partial f}{\partial x}$
- $\frac{\partial f}{\partial y}$ or $f_y$: The partial of $f$ with respect to $y$, written as  $\frac{\partial f}{\partial y}$

**How to compute partial derivatives: a step-by-step guide**

1. Identify the function $f$ and the specific variable you need to differentiate with respect to  

2. Hold one variable constant and take derivatives as usual. Ignore other variables and take them as constants

3. Apply the standard rules of differentiation, such as **power rule, product rule, chain rule** to the variable you are differentiating with respect to

**Interpretation:**                                                                   
- Partial Derivatives tells us how quickly the function's value changes when you only change the input, while keeping the inputs fixed

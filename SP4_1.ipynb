$$\frac{dy}{dx} = x e^{5x} y^2$$

**Solution:**

$$\int y^{-2} dy = \int x e^{5x} dx$$

$$\int y^{-2} dy = -y^{-1} + C_1 = -\frac{1}{y} + C_1$$

$u = x \implies du = dx$$dv = e^{5x} dx \implies v = \frac{1}{5} e^{5x}$$$\int x e^{5x} dx = x \left(\frac{1}{5} e^{5x}\right) - \int \frac{1}{5} e^{5x} dx$$$$= \frac{1}{5} x e^{5x} - \frac{1}{5} \left(\frac{1}{5} e^{5x}\right) + C_2$$$$\int x e^{5x} dx = \frac{1}{5} x e^{5x} - \frac{1}{25} e^{5x} + C_2$$

$$-\frac{1}{y} = \frac{1}{5} x e^{5x} - \frac{1}{25} e^{5x} + C$$

$$\frac{1}{y} = -\frac{1}{5} x e^{5x} + \frac{1}{25} e^{5x} - C$$

$$\frac{1}{y} = \frac{1}{25} e^{5x} - \frac{1}{5} x e^{5x} + K$$

$$y = \frac{1}{\frac{1}{25} e^{5x} - \frac{1}{5} x e^{5x} + K}$$

**Problem 14:**

Solve the initial-value problem:
$
\frac{dy}{dx} = x\,y^{2}, \qquad y(0) = 1.
$

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Step 1 — Separate variables**

Move all $(y)$-terms to one side and all $(x)$-terms to the other:
    
$
\frac{dy}{dx} = x y^{2}
\quad\Longrightarrow\quad
\frac{dy}{y^{2}} = x\,dx.
$


**Step 2 — Integrate both sides**

$
\int \frac{dy}{y^{2}} = \int x\,dx.
$

Compute each integral:

$
-\,\frac{1}{y} = \frac{x^{2}}{2} + C.
$


**Step 3 — Solve for $(y)$**
                 
$
-\,\frac{1}{y} = \frac{x^{2}}{2} + C
\quad\Longrightarrow\quad
\frac{1}{y} = -\frac{x^{2}}{2} - C.
$

Rename the constant $(C_1 = -C)$:
                 
$
\frac{1}{y} = C_1 - \frac{x^{2}}{2}
\quad\Longrightarrow\quad
y(x) = \frac{1}{\,C_1 - \dfrac{x^{2}}{2}\,}.
$


**Step 4 — Apply initial condition $(y(0)=1)$**

$
1 = \frac{1}{C_1}
\quad\Longrightarrow\quad
C_1 = 1.
$

Solution is:
$
\boxed{\,y(x) = \frac{1}{\,1 - \dfrac{x^{2}}{2}\,}\,}.
$


**Step 5 — Domain of the solution**

The denominator must be nonzero:

$
1 - \frac{x^{2}}{2} \neq 0
\quad\Longrightarrow\quad
x \neq \pm\sqrt{2}.
$

Solution is valid on:
$
-\sqrt{2} < x < \sqrt{2}.
$


**Step 6 — Verification**
    
Differentiate:
$
y(x) = \left(1 - \frac{x^{2}}{2}\right)^{-1}.
$

$
y'(x)
= x\left(1 - \frac{x^{2}}{2}\right)^{-2}.
$

Compute the right-hand side:

$
x\,y^{2} = x\left(1 - \frac{x^{2}}{2}\right)^{-2}.
$

Since $(y' = x y^{2})$, the differential equation is satisfied.

**Final Answer**
$
\boxed{y(x) = \frac{1}{1 - \dfrac{x^{2}}{2}}}
\quad \text{for} \quad -\sqrt{2} < x < \sqrt{2}.
$

\frac{1}{y - 1}\,dy = x^2\,dx.

y = 1 + K e^{x^3/3}.

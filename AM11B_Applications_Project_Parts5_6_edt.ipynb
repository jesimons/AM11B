## Section 5.1: Profit Maximization

### Background

A firm's **profit** is revenue minus costs, which can be modeled as:

$$\pi(L, K) = pQ - wL - rK$$

where:
- $p$ = price of output
- $Q = f(L, K)$ = production function
- $L$ = labor (number of workers)
- $K$ = capital expenditures (rent, utilities, etc.)
- $w$ = wage (cost per worker)
- $r$ = rental rate (cost per unit of capital)

Here, revenue is $pQ$ (price times quantity sold) and costs include total wages $wL$ and total capital expenditures $rK$.

### Finding the Optimal Inputs

To maximize profit, we take partial derivatives and set them equal to zero:

$$\frac{\partial \pi}{\partial L} = p \cdot MP_L - w = 0$$
$$\frac{\partial \pi}{\partial K} = p \cdot MP_K - r = 0$$



**Economic Interpretation:** At the optimum, the value of the marginal product (p·MP) equals the input cost. This makes sense: keep hiring until the last worker's contribution equals their wage.

We should also check second-order conditions to make sure we have a maximum, not a minimum!

## Section 6.1: Utility Maximization with Budget Constraint

### Background

Consumers want to maximize their satisfaction (utility) but face a budget constraint, which limits their choices to bundles they can afford with their income. This is the fundamental problem of consumer choice.

**The Problem:**

Maximize: $U(x_1, x_2)$

Subject to: $p_1 x_1 + p_2 x_2 = I$

where:
- $U(x_1,x_2)$ = utility function
- $x_1$ = quantity of good 1 (like bananas)
- $x_2$ = quantity of good 2 (like apples)
- $p_1, p_2$ = prices of goods 1 and 2
- $I$ = consumer's income
- The budget constraint says: total spending on both goods cannot exceed income.

### Solution Method 1: Tangency Condition

At the optimum, the slope of the **indifference curve** (see Phase 1 of project) equals the slope of the budget line:

$$MRS = \frac{p_1}{p_2}$$

or equivalently:

$$\frac{MU_{x_1}}{MU_{x_2}} = \frac{p_1}{p_2}$$

**Economic intuition:** The consumer's willingness to trade (MRS) should equal the market's trading rate (price ratio).

### Example: Applying the Tangency Condition

Suppose a consumer has the following utility function:

$U(x_1, x_2) = x_1^{1/2} x_2^{1/2}$

Prices and income are:

- $p_1 = 2$  
- $ p_2 = 4$  
- $ I = 100$ 



### **Step 1: Compute the MRS**

Marginal utilities:

$ MU_{x_1} = \frac{1}{2} x_1^{-1/2} x_2^{1/2}$ , 
$ \qquad $ 
$ MU_{x_2} = \frac{1}{2} x_1^{1/2} x_2^{-1/2}$ 


The marginal rate of substitution (MRS) is: $ MRS = \frac{MU_{x_1}}{MU_{x_2}} = \frac{x_2}{x_1}$ 




### **Step 2: Apply the Tangency Condition**

At the optimum: $ MRS = \frac{p_1}{p_2}$ 


Plug in values: $ \frac{x_2}{x_1} = \frac{2}{4} = \frac{1}{2}$ 


So the optimal relationship between the goods is: $ x_2 = \frac{1}{2} x_1$ 




### **Step 3: Use the Budget Constraint**

Budget constraint: $ 2x_1 + 4x_2 = 100$ 


Substitute $ x_2 = \tfrac{1}{2} x_1$ :


$ 2x_1 + 4\left(\tfrac{1}{2} x_1\right)$ 
$ = 2x_1 + 2x_1$ 
$ = 4x_1$ 
$ = 100$ 


Thus: $ x_1 = 25$ 


Then: $ x_2 = \tfrac{1}{2}(25) = 12.5$

### Solution Method 2: Lagrange Multipliers

Set up the **Lagrangian**:

$$\mathcal{L}  = U(x_1, x_2) - \lambda(p_1 x_1 + p_2 x_2 - I)$$

Take three partial derivatives and set equal to zero:
1. $\frac{\partial \mathcal{L} }{\partial x_1} = MU_{x_1} - \lambda p_1 = 0$
2. $\frac{\partial \mathcal{L} }{\partial x_2} = MU_{x_2} - \lambda p_2 = 0$
3. $\frac{\partial \mathcal{L} }{\partial \lambda} = I - p_1 x_1 - p_2 x_2 = 0$

Then solve this system of three equations for $x_1, x_2,$ and $\lambda$.

**The meaning of $\lambda$:** The Lagrange multiplier $\lambda$ represents the marginal utility of income - how much extra utility you'd get from one more dollar of income.

### Example: Solving Utility Maximization with the Lagrangian Method

We use the same utility function, prices, and income as before:


$U(x_1, x_2) = x_1^{1/2} x_2^{1/2}$, $\quad$ $p_1 = 2,\; p_2 = 4,\; I = 100$.



### **Step 1: Set Up the Lagrangian**


$\mathcal{L} = x_1^{1/2} x_2^{1/2}$ 
$- \lambda(2x_1 + 4x_2 - 100)$




### **Step 2: Take the First-Order Conditions (FOCs)**

1. **FOC for $x_1$:**


$\frac{\partial \mathcal{L}}{\partial x_1}$
$= \frac{1}{2} x_1^{-1/2} x_2^{1/2} - 2\lambda = 0$


2. **FOC for $x_2$:**


$\frac{\partial \mathcal{L}}{\partial x_2}$
$= \frac{1}{2} x_1^{1/2} x_2^{-1/2} - 4\lambda = 0$


3. **FOC for $lambda$:** (the budget constraint)


$2x_1 + 4x_2 = 100$




### **Step 3: Use the First Two FOCs to Eliminate $\lambda$**

From the first FOC:


$\lambda = \frac{1}{4} x_1^{-1/2} x_2^{1/2}$


From the second FOC:


$\lambda = \frac{1}{8} x_1^{1/2} x_2^{-1/2}$


Set them equal:


$\frac{1}{4} x_1^{-1/2} x_2^{1/2}$
$=$
$\frac{1}{8} x_1^{1/2} x_2^{-1/2}$


Multiply both sides by 8:


$2\, x_1^{-1/2} x_2^{1/2} = x_1^{1/2} x_2^{-1/2}$


Rewrite:


$2 \cdot \frac{x_2}{x_1} = 1$


So:


$\frac{x_2}{x_1} = \frac{1}{2}$


Thus:


$x_2 = \frac{1}{2} x_1$.



### **Step 4: Substitute into the Budget Constraint**

Budget: $2x_1 + 4x_2 = 100$


Substitute $x_2 = \tfrac{1}{2}x_1$:


$2x_1 + 4\left(\tfrac{1}{2}x_1\right)$
$= 2x_1 + 2x_1$
$= 4x_1$
$= 100$

Solve:


$x_1 = 25$


And:


$x_2 = \tfrac{1}{2}(25) = 12.5$

## Section 6.2: Cost Minimization

### Background

Firms also face a constrained optimization problem: minimize the cost of producing a target level of output.

**The Problem:**

Minimize: $C = wL + rK$ (total cost)

Subject to: $f(L, K) = Q_0$ (must produce at least $Q_0$ units)

### Solution: Tangency Condition

At the optimum:

$$\frac{MP_L}{MP_K} = \frac{w}{r}$$

**Economic interpretation:** The ratio of marginal products should equal the ratio of input prices. If MP<sub>L</sub> /w > MP<sub>K</sub> /r, the firm should use more labor and less capital.

### Lagrangian Approach 

$$  \mathcal{L} = wL + rK - \lambda(f(L, K) - Q_0)$$

First-order conditions:
1. $\frac{\partial \mathcal{L} }{\partial L} = w - \lambda \cdot MP_L = 0$
2. $\frac{\partial \mathcal{L} }{\partial K} = r - \lambda \cdot MP_K = 0$
3. $\frac{\partial \mathcal{L} }{\partial \lambda} = Q_0 - f(L,K) = 0$

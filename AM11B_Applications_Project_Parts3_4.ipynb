## Section 3.1: Marginal Functions

One of the more basic examples of how economists may use calculus is the concept of **marginal functions**.

### What Does "Marginal" Mean in Economics?

In economics, "marginal" means **the additional amount** resulting from a one-unit increase in some variable. Mathematically, this is just the **derivative**!

For example:
- **Marginal Cost (MC)** = the cost of producing one more unit
- **Marginal Revenue (MR)** = the revenue from selling one more unit  
- **Marginal Product** = the additional output from one more unit of input

If we have a total cost function $C(q)$ where $q$ is quantity produced, then:

$$MC(q) = \frac{dC}{dq}$$

### Example: A Simple Cost Function

Suppose a firm's total cost of producing $q$ units is:

$$C(q) = 100 + 10q + 0.5q^2$$

This includes:
- **Fixed costs**: $100$ (costs even when $q=0$)
- **Variable costs**: $10q + 0.5q^2$ (costs that depend on quantity)

Let's visualize this and find the marginal cost function. Run the code below.

```python
def total_cost(q):
    """
    Calculate total cost for producing q units.
    C(q) = 100 + 10q + 0.5q^2
    """
    return 100 + 10*q + 0.5*q**2

def marginal_cost(q):
    """
    Calculate marginal cost at quantity q.
    MC(q) = dC/dq = 10 + q
    """
    # Take the derivative of C(q) = 100 + 10q + 0.5q^2
    MC = 10+q
    return MC

# Test the functions
q_test = 20
print(f"At q = {q_test}:")
print(f"  Total Cost: C({q_test}) = ${total_cost(q_test):.2f}")
print(f"  Marginal Cost: MC({q_test}) = ${marginal_cost(q_test):.2f}")
print(f"\nInterpretation: The {q_test}th unit costs approximately ${marginal_cost(q_test):.2f} to produce.")
```

Notice that the interpretation is about how much it costs to produce one additional unit (in this case, the 20th unit).

We can also visualize cost versus marginal cost by running the code below.

```python
# Visualize total and marginal cost
q_values = np.linspace(0, 50, 100)
C_values = [total_cost(q) for q in q_values]
MC_values = [marginal_cost(q) for q in q_values]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot total cost
ax1.plot(q_values, C_values, 'b-', linewidth=2.5)
ax1.set_xlabel('Quantity (q)', fontsize=11)
ax1.set_ylabel('Total Cost ($)', fontsize=11)
ax1.set_title('Total Cost Function: C(q) = 100 + 10q + 0.5q²', fontsize=12, weight='bold')
ax1.grid(True, alpha=0.3)

# Plot marginal cost
ax2.plot(q_values, MC_values, 'r-', linewidth=2.5)
ax2.set_xlabel('Quantity (q)', fontsize=11)
ax2.set_ylabel('Marginal Cost ($)', fontsize=11)
ax2.set_title('Marginal Cost Function: MC(q) = 10 + q', fontsize=12, weight='bold')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Section 3.2 Price-Elasticity of Demand

**Price-Elasticity of Demand** refers to a ratio of percentage changes in demand to prices. For instance, if you raise the price of a burrito by 10%, and the demand changes by 20%, the price-elasticity of demand is:
$$\frac{20 \%}{10 \%} = 2$$
The higher this ratio, the more sensitive the market is to price changes. The lower this ratio, the less sensitive (e.g. less **elastic**) the market is.  

When a product is a necessity (like food), elasticity tends to be low. Grocery staples like milk have low price elasticity. For example,  if the price of a gallon of milk rises from $ \$3.50$ to $ \$3.70$, most families still buy about the same amount because they consider it essential.


If a specialty snack gets more expensive at a regular supermarket, a shopper might buy a smaller amount, wait to find it cheaper at Grocery Outlet, or skip it entirely. For example, a shopper usually buys a specialty snack at Grocery Outlet for $ \$2.00$ a bag and picks up four bags a month. When they see the same snack priced at $ \$5.50$ at a regular supermarket, they buy none and wait to get it at Grocery Outlet.


Since we have used the word "change" above, we can capture this concept with this formula for elasticity:
$$ E_{q/p} = \frac{\% \Delta q}{\% \Delta p}= \frac{\frac{\Delta q}{q}}{\frac{\Delta p}{p}} = \frac{\Delta q}{\Delta p}\frac{p}{q}$$
The formula above is written for **average elasticity**. To bring this into the realm of calculus, we have the formula:

$$ \eta_{q/p} =  \frac{d q}{d p}\frac{p}{q}$$

This formula above is called **point elasticity** and is the elasticity formula we will use.

## Section 4.1: Marginal Products in a Multivariable World

### Background

Just as when we work with functions of one variable, the **marginal product** of an input tells us how much additional output we get from one more unit of that input. However, if we have multiple inputs (a *multivariable function*), we hold all other inputs constant.

**Marginal Product of Labor (MPL):**
$$MP_L = \frac{\partial Q}{\partial L}$$

Here $Q$ is production and $L$ is labor. This measures the extra output from hiring one more worker, keeping capital fixed.

**Marginal Product of Capital (MPK):**
$$MP_K = \frac{\partial Q}{\partial K}$$

Here $Q$ is production and $K$ is capital. This measures the extra output from one more machine, keeping labor fixed.

**Diminishing Marginal Returns:** Typically, as we add more and more of one input (holding others fixed), the marginal product decreases. The first worker adds a lot, but the 100th worker adds less.

### Marginal Returns

A key property of production functions is **diminishing marginal returns**: as you add more of one input (holding others constant), each additional unit contributes less and less to output.

**Task:** Run the code below to demonstrate diminishing marginal returns to capital for the Cobb-Douglas equation. Calculate MPK for increasing levels of capital while holding labor constant.

```python
def marginal_product_capital(A, K, L, alpha):
    """
    Calculate the marginal product of capital.

    Returns:
    float: MPK = ∂Y/∂K
    """
    # MPK
    MPK = A*alpha*K**alpha*(1-alpha)*L**-alpha
    return MPK

def marginal_product_labor(A, K, L, alpha):
    """
    Calculate the marginal product of labor.

    Returns:
    float: MPL = ∂Y/∂L
    """
    # MPL
    MPL = A*K**alpha*(1-alpha)*L**-alpha
    return MPL

# Fixed parameters
A = 1
L_fixed = 100
alpha = 0.3

# Calculate MPK for different levels of capital
K_values = np.array([50, 100, 150, 200, 250, 300])
MPK_values = np.array([marginal_product_capital(A, K, L_fixed, alpha) for K in K_values])

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(K_values, MPK_values, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Capital (K)', fontsize=12)
plt.ylabel('Marginal Product of Capital (MPK)', fontsize=12)
plt.title('Diminishing Marginal Returns to Capital', fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()

# Print the values
print("\nCapital (K) | MPK")
print("-" * 25)
for K, MPK in zip(K_values, MPK_values):
    print(f"{K:11.0f} | {MPK:.6f}")
```

## Section 4.2: Marginal Utility

### Background

The **marginal utility** of a good measures how much additional satisfaction a consumer gets from consuming one more unit of that good.

$$MU_x = \frac{\partial U}{\partial x}, \quad MU_y = \frac{\partial U}{\partial y}$$

### Marginal Rate of Substitution (MRS)

The **MRS** tells us the rate at which a consumer is willing to trade one good for another while staying equally happy:

$$MRS = \frac{MU_y}{MU_x}$$

**Interpretation:** If MRS = 2, the consumer is willing to give up 2 units of x to get 1 more unit of y (and remain equally satisfied).

## Section 4.3: Returning to Cobb-Douglas

Recall that the **Cobb-Douglas production function** is:

$$Y = A K^\alpha L^\beta$$

where:
- $Y$ is output (total production)
- $K$ is capital (machinery, buildings, equipment)
- $L$ is labor (number of workers or hours worked)
- $A$ is total factor productivity (technology level)
- $\alpha$ and $\beta$ are the output elasticities of capital and labor

When $\alpha + \beta = 1$, the function exhibits **constant returns to scale**, meaning if you double all inputs, you double the output.

Notice the word *elasticities* shows up here! Let's explore this a little bit.

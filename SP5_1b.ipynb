**Problem 10:**

Consider the function $g(s,t) = s^2 + t^2$.

1. Find the level curves for $c = 0, 1, 4, 9$.
2. Sketch or describe what these level curves look like.
3. What do these level curves tell you about the 3D graph of g(s,t)?
**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part 1: Find the level curves**

A level curve is found by setting $g(s,t) = c$ for a constant c.

For $c = 0$:
$$s^2 + t^2 = 0$$
This is only satisfied when $s = 0$ and $t = 0$, so the level curve is a single point at the origin.

For $c = 1$:
$$s^2 + t^2 = 1$$
This is a circle centered at the origin with radius 1.

For $c = 4$:
$$s^2 + t^2 = 4$$
This is a circle centered at the origin with radius 2.

For $c = 9$:
$$s^2 + t^2 = 9$$
This is a circle centered at the origin with radius 3.

**Part 2: Describe the level curves**

The level curves are concentric circles (circles with the same center) centered at the origin. As c increases, the radius of the circles increases. Specifically, the radius is $r = \sqrt{c}$.

**Part 3: What do these tell us about the 3D graph?**
The level curves tell us that:

- The 3D graph of $g(s,t) = s^2 + t^2$ is a paraboloid (bowl shape) opening upward.
- The lowest point (minimum) is at the origin (0,0) where g = 0.
- As you move away from the origin in any direction, the function value increases.
- The function has circular symmetry around the vertical axis through the origin.
- The surface gets steeper as you move away from the origin (since the circles get farther apart for equal increases in height).

**Solution:**

Start with the level-curve equation:
x^2 - 4y + 2xy = 6

Rearrange the equation to isolate y terms:
2xy - 4y = 6 - x^2

Factor out y:
y(2x - 4) = 6 - x^2

Solve for y (assuming x ≠ 2):
y = (6 - x^2) / (2x - 4)

This equation describes the level curve in the xy-plane.
It is not a line or a circle; it is a rational curve because y depends on x through a fraction involving x^2.

When x = 2, the denominator becomes 0, so no finite y satisfies the equation.
This means the level curve has a vertical asymptote at x = 2.

On a 3D surface, the function f(x,y) defines a surface z = f(x,y).
A level curve is a horizontal “slice” of this surface at a fixed height.
Setting f(x,y) = 6 means we slice the surface at z = 6.
The curve y = (6 - x^2) / (2x - 4) shows exactly where the surface touches the height z = 6.
In 3D, this appears as the intersection of the surface with the horizontal plane z = 6.
In 2D (the xy-plane), we see only the shadow of that intersection, which is why we get the curve above.

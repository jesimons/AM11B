## Section 1.1: Consumer and Producer Surplus

### Background

In a market, consumers and producers both benefit from trade:

- **Consumer Surplus**: The difference between what consumers are willing to pay and what they actually pay. This measures the benefit consumers receive from trade.

- **Producer Surplus**: The difference between what producers receive and the minimum they would accept. This measures the benefit producers receive from trade.

We use integration to calculate these quantities because we need to sum up the surplus for each unit traded.

### The Setup

Consider a market with:
- **Demand curve** $D(q)$: Shows the price consumers are willing to pay for quantity $q$
- **Supply curve** $S(q)$: Shows the price producers need to receive for quantity $q$
- **Equilibrium**: Occurs where $D(q) = S(q)$, giving equilibrium price $p^*$ and quantity $q^*$

### The Formulas

**Consumer Surplus:**
$$CS = \int_0^{q^*} [D(q) - p^*] \, dq$$

This represents the area between the demand curve and the price line.

**Producer Surplus:**
$$PS = \int_0^{q^*} [p^* - S(q)] \, dq$$

This represents the area between the price line and the supply curve.

**Total Surplus** = Consumer Surplus + Producer Surplus

## Section 1.2: Gini Coefficient and Income Inequality

### Background

The **Gini coefficient** measures income inequality in a population. It ranges from 0 (perfect equality) to 1 (perfect inequality).

The calculation uses the **Lorenz curve**, which is a graph showing:
- **x-axis**: Cumulative share of population (from poorest to richest)
- **y-axis**: Cumulative share of income

For example, the point (0.4, 0.15) on the Lorenz curve means "the bottom 40% of the population earns 15% of total income."

**Properties of Lorenz curves:**
- Always passes through (0,0) and (1,1)
- Always below or on the line y = x (the line of perfect equality)
- Is always increasing and concave up

### The Formula

Given a Lorenz curve $L(x)$, the Gini coefficient is:

$$G = 1 - 2\int_0^1 L(x) \, dx$$

Geometrically, $G$ equals the area between the line of equality and the Lorenz curve, divided by the total area under the equality line (which is 0.5).

**Interpretation:**
- $G = 0$: Perfect equality (everyone has the same income)
- $G = 0.3$: Relatively low inequality
- $G = 0.5$: High inequality
- $G = 1$: Perfect inequality (one person has all income)

## Section 2.1: Production Functions

### Background

A **production function** shows how much output a firm can produce from different combinations of inputs. The most common form is:

$$Q = f(L, K)$$

where:
- $Q$ = output (quantity produced)
- $L$ = labor input (number of workers or hours)
- $K$ = capital input (machines, equipment, buildings)

**Example:** A bakery might have $Q = 10\sqrt{L \cdot K}$, meaning output depends on both the number of workers and the number of ovens.

### Common Production Function: Cobb-Douglas

One widely-used production function is:

$$Q = AL^{\alpha}K^{\beta}$$

where $A$, $\alpha$, and $\beta$ are positive constants. This is called a **Cobb-Douglas production function**.

### Returns to Scale

If we double all inputs, what happens to output?

- **Constant returns to scale**: Output exactly doubles
- **Increasing returns to scale**: Output more than doubles (economies of scale)
- **Decreasing returns to scale**: Output less than doubles

## Section 2.2: Utility Functions

### Background

A **utility function** represents a consumer's preferences over different "bundles" of goods. If we have two goods x and y, specific values of (x,y) are called a bundle. Then, a utility function looks like:

$$U = u(x, y)$$

The utility function assigns a number to each combination ("bundle") of goods, where higher numbers mean the consumer prefers that bundle more.

**Example:** $U = xy$ means the consumer values having some of both goods (x and y multiply), rather than all of one good.

### A Common Utility Function
consider a utility function over bundles of bananas (B) and apples (A)

$$
U(B, A) = B^{\alpha}A^{1-\alpha}
$$

where $ \alpha \in [0,1] $. We can re-write this in more common notation as:

$$U(x_1, x_2) = x_1^\alpha x_2^{1-\alpha}$$

where $x_1$ and $x_2$ are quantities of two goods, and $\alpha \in [0,1]$ is a parameter that represents the consumer's preference for good 1 relative to good 2.

### Question 2.1: Implementing the Utility Function

Let's visualize this utility function and understand its behavior, at least in one dimension first. To do this, we will set $x_1 = 2$ (assume it's held constant) and keep $\alpha = \frac{1}{3}$. Run the following code to see what $U$ looks like.

```python
def utility(x1, x2, alpha):
    """
    Calculate utility from consuming x1 units of good 1 and x2 units of good 2.

    Inputs:
    x1: Quantity of good 1
    x2: Quantity of good 2

    Parameters:
    alpha: Preference parameter (between 0 and 1)

    Output:
    Utility level
    """
    return x1**alpha * x2**(1-alpha)

fig, ax = plt.subplots()
x1 = 2
x2 = np.linspace(0, 10, 100)
ax.plot(x2, utility(x1,x2,1/3))
ax.set_xlabel("$x_2$")
ax.set_ylabel("U($x_1$=2, $x_2$)")
```

```python
def indifference_curve(x1, ubar, alpha):
    """
    Calculate x2 values that give utility level ubar for given x1 values.

    Parameters:
    x1 (array): Quantities of good 1
    ubar (float): Target utility level
    alpha (float): Preference parameter

    Returns:
    array: Corresponding x2 values
    """

    x2 = (ubar/x1**alpha)**(1/(1-alpha)) # EDIT HERE
    return x2

# Plot indifference curves
x1_vals = np.linspace(1, 50, 100) # EDIT HERE (1= start, 50 = stop, may want to change)
alpha = 0.5
utility_levels = [5, 10, 15, 20, 25, 30, 35] # EDIT HERE (may want to change levels)

plt.figure(figsize=(10, 8))
for U in utility_levels:
    x2_vals = indifference_curve(x1_vals, U, alpha)
    plt.plot(x1_vals, x2_vals, linewidth=2, label=f'U = {U}')

plt.xlabel('Good 1 ($x_1$)', fontsize=12)
plt.ylabel('Good 2 ($x_2$)', fontsize=12)
plt.title('Indifference Curves', fontsize=14)
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(0, 50)
plt.ylim(0, 50)
plt.show()
```

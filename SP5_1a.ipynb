**Problem 1:**

$$f(x, y) = \frac{\sqrt{x - y}}{x}$$

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

The domain is restricted by two conditions:Square Root: The expression under the square root must be non-negative: $x - y \geq 0 \implies y \leq x$.Fraction: The denominator cannot be zero: $x \neq 0$.The domain $D$ is the set of all points $(x, y)$ such that:$$D = \{(x, y) \in \mathbb{R}^2 \mid y \leq x \text{ and } x \neq 0\}$$

To find the range, let $z = f(x, y) = \frac{\sqrt{x - y}}{x}$. Since the numerator $\sqrt{x-y}$ is always non-negative (zero or positive) and the denominator $x$ must be non-zero, the sign of $z$ depends only on $x$.From the domain $y \leq x$, we can rewrite the function as:$$z = \frac{\sqrt{x - y}}{x}$$Since $x = y + (x-y)$, and $x-y$ is the quantity under the square root, let $u = \sqrt{x-y}$ (so $u \geq 0$). Then $x = y + u^2$.

*(b)* Determine the *range* of $(f(x,y))$.

*Solution:*

<details>
<summary><b>Click to reveal solution</b></summary>

 Finding the Domain

The function has two expressions that impose restrictions:

1. $9 - x^2 - y^2 \ge 0 \quad\Rightarrow\quad x^2 + y^2 \le 9.$

2. The natural log requires  $x + y + 5 > 0 \quad\Rightarrow\quad x + y > -5$


Thus, the *domain* is the set of all $((x,y))$ inside or on the circle $(x^2 + y^2 \le 9)$ and on the side of the line $(x+y>-5)$


Finding the Range

- The square root part satisfies  
  
  $0 \le \sqrt{9 - x^2 - y^2} \le 3$
  

- The logarithm satisfies  
  
  $\ln(x+y+5)$
  
  where:
  - its minimum approaches $(-\infty)$ as $(x+y+5 \to 0^+)$,
  - its maximum occurs when $(x+y)$ is maximized inside the disk.

Inside the disk $(x^2 + y^2 \le 9)$, the maximum value of $(x+y)$ occurs along the line $(x=y)$, giving  

$x=y=\frac{3}{\sqrt{2}}\quad\Rightarrow\quad x+y = \frac{6}{\sqrt{2}} = 3\sqrt{2}$.


Thus the maximum of the log term is  

$\ln(3\sqrt{2}+5)$


Adding the maximum of the square root (which is 3), the overall maximum of (f) is  

$f_{\max} = 3 + \ln(3\sqrt{2}+5)$


Since the log term can go to $(-\infty)$,  

$f_{\min} = -\infty$


Final Range

$(-\infty,\; 3 + \ln(3\sqrt{2}+5)]$.

**Problem 10:**

Consider the function $f(x,y) = \sqrt{9 - x^2 - y^2}$.

1. Find the domain of f(x,y).
2. Describe the domain geometrically.
3. Find the range of f(x,y).
4. Evaluate f(1,2) if it exists.
**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**Part 1: Find the domain**

For the square root to be defined, we need:
$$9 - x^2 - y^2 \geq 0$$
$$x^2 + y^2 \leq 9$$

**Domain**: $\{(x,y) : x^2 + y^2 \leq 9\}$

**Part 2: Describe the domain geometrically**

The domain is a disk (circle and its interior) centered at the origin with radius 3. This includes all points on and inside the circle $x^2 + y^2 = 9$.

**Part 3: Find the range**

Since $x^2 + y^2 \leq 9$, we have:
$$9 - x^2 - y^2 \geq 0$$
$$\sqrt{9 - x^2 - y^2} \geq 0$$

The maximum value occurs when $x^2 + y^2 = 0$ (at the origin):
$$f(0,0) = \sqrt{9 - 0 - 0} = 3$$

The minimum value occurs when $x^2 + y^2 = 9$ (on the boundary):
$$f(x,y) = \sqrt{9 - 9} = 0$$

**Range**: $[0, 3]$

**Part 4: Evaluate f(1,2)**

First check if (1,2) is in the domain:
$$1^2 + 2^2 = 1 + 4 = 5 \leq 9$$ ✓

So (1,2) is in the domain. Now evaluate:
$$f(1,2) = \sqrt{9 - 1 - 4} = \sqrt{4} = 2$$
</details>

**Problem 31:**

Consider the function
$$f(x,y) = \sqrt{16 - x^2 - y^2}.$$

a) Find the domain of $f$.  
b) Find the range of $f$.  
c) Is the point $(3,1)$ in the domain? What about $(4,3)$?

**Solution:**

<details>
<summary><b>Click to reveal solution</b></summary>

**a) Domain**

The expression inside the square root must be nonnegative:

$$16 - x^2 - y^2 \ge 0 \quad\Longrightarrow\quad x^2 + y^2 \le 16.$$

So the domain is all points $(x,y)$ in the plane with $x^2 + y^2 \le 16$.  
Geometrically, this is the disk of radius 4 centered at the origin (including the boundary circle).

---

**b) Range**

Inside the square root we have $16 - x^2 - y^2$.

From the domain condition $x^2 + y^2 \le 16$, the smallest value of $16 - x^2 - y^2$ is $0$ (when $x^2 + y^2 = 16$), and the largest value is $16$ (when $x^2 + y^2 = 0$).

So:

- Minimum inside the root: $0 \Rightarrow f(x,y) = \sqrt{0} = 0$.  
- Maximum inside the root: $16 \Rightarrow f(x,y) = \sqrt{16} = 4$.

Therefore, the range is

$$0 \le f(x,y) \le 4,$$

or in interval notation: $[0,4]$.

---

**c) Checking specific points**

- For $(3,1)$:

  $$3^2 + 1^2 = 9 + 1 = 10 \le 16,$$

  so $(3,1)$ is in the domain.

- For $(4,3)$:

  $$4^2 + 3^2 = 16 + 9 = 25 > 16,$$

  so $(4,3)$ is not in the domain (it would make the square root negative).

$$ f(x,y) = \ln(4 - x^{2} - y) $$

**The natural logarithm is only defined when its input is strictly positive.**
    
**So we require**


$$ 4 - x^{2} - y > 0 $$


**This inequality tells us which points $(x,y)$ are allowed.**


**Step 2: Solve the inequality (find the domain)**

**Starting from**

$$ 4 - x^{2} - y > 0 $$

**solve for $y$:**


$$ -x^{2} - y > -4 $$


$$ -y > -4 + x^{2} $$


**Multiply both sides by $-1$ (and reverse the inequality):**


$$ y < 4 - x^{2} $$


**Therefore, the domain is:**
    
$$ (x,y) in \mathbb{R}^{2} : y < 4 - x^{2}\} $$


**This means all points below the parabola $y = 4 - x^{2}$.**



**Step 3: Find the range**

**Let:**

$$ u = 4 - x^{2} - y. $$

**From the domain, we know that $u>0$.**

**As $y$ approaches the boundary $y = 4 - x^{2}$ from below,**


**As** $$y \to -\infty $$


$$ u = 4 - x^{2} - y \to +\infty \quad \text{and} \quad \ln(u) \to +\infty. $$


**So the function takes every real value.**

**Thus, the range is**

$$ (-\infty,\, \infty) $$

**Final Answers:**

**Domain:** $$ \{(x,y)\in\mathbb{R}^{2} : y < 4 - x^{2}\} $$


**Range:** $$(-\infty,\, \infty) $$

**Problem 84:**

Consider the function  
$$f(x,y) = x^2 + 2y.$$

1. Identify the domain of this function.  
2. Identify the range of this function.  

Solution:

<details>
<summary><b>Click to reveal solution</b></summary>

Step 1: Domain

The function is  
$$f(x,y) = x^2 + 2y.$$

There are no restrictions on $x$ or $y$:

- We can plug in any real number for $x$.  
- We can plug in any real number for $y$.  

So the domain is all pairs $(x,y)$ of real numbers:

$$\text{Domain} = (-\infty,\infty)\times(-\infty,\infty)$$

We write the domain as $(-\infty,\infty)\times(-\infty,\infty)$ because the function has two input variables. Each interval $(-\infty,\infty)$ represents all possible real values for one variable. The $\times$ symbol indicates that we take all possible pairs $(x,y)$ from both intervals, since $x$ and $y$ can vary independently.

Step 2: Range

The output is  
$$f(x,y) = x^2 + 2y.$$

- The term $x^2$ is always greater than or equal to 0.  
- The term $2y$ can be any real number, positive or negative.

Since we can make $2y$ very large or very negative, and $x^2$ only adds nonnegative values, the function can produce any real number.

So the range is:

$$\text{Range} = (-\infty,\infty).$$

Conclusion

For this multivariable function:

- The domain is every possible pair of real numbers.  
- The range is every possible real output.

This function is simple but shows how multivariable inputs can produce a wide variety of outputs.
</details>
